## Concept focus — Evaluation metrics, calibration, and disciplined experimentation

A model is only good relative to the decision it supports. This module matters because accuracy alone can hide failure, especially under imbalance, bad calibration, or unstable validation design.

```text
train -> validate -> inspect metrics -> change one thing -> repeat

wrong metric
=> wrong optimization target
```

### How to think about it
Choose metrics by consequence, not habit. Ask what error hurts most, how thresholds change outcomes, and whether your evaluation setup matches how the model will actually be used in production.
