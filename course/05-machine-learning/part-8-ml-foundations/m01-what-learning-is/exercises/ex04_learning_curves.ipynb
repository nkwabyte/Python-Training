## Concept focus — Problem framing, splits, baselines, and the learning loop

Machine learning starts long before model choice. You need to understand what the target is, what the data can reveal, how to split it safely, and what baseline performance means before any training run is worth trusting.

```text
define problem -> prepare split -> build baseline -> train model -> evaluate -> iterate

bad framing at the start
=> wasted experiments later
```

### How to think about it
Think like an experimental scientist. Every notebook in this module is about reducing avoidable confusion: leakage, bad splits, weak baselines, and stories that seem convincing until the held-out data disagrees.
