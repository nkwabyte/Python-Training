## Concept focus — Self-attention, masking, positional information, and scaling cost

The transformer is central to modern ML because it replaces recurrence with parallel attention over token relationships. Understanding its cost and structure is essential for both language modeling and many multimodal systems.

```text
tokens -> embeddings + positions -> self-attention -> feedforward -> next block
                               |
                           masks control what can be seen
```

### How to think about it
Treat attention as content-based routing: every token asks what other tokens matter right now. Then ask what this flexibility costs in memory, compute, and sequence length — that is where practical engineering trade-offs begin.
