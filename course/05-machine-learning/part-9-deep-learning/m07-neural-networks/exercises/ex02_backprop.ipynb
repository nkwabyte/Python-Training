## Concept focus — Nonlinearity, backpropagation, initialization, and optimizers

A neural network only becomes more expressive than a linear model when its layers, activations, and gradients interact correctly. This module is about the mechanics that make multilayer models trainable rather than decorative.

```text
inputs -> linear -> nonlinearity -> linear -> output -> loss
             ^                             |
             |-------- backpropagation ----|
```

### How to think about it
Picture every layer as a transformation that must preserve useful signal in both directions: forward activations and backward gradients. If either collapses, training stalls no matter how good the dataset looks.
