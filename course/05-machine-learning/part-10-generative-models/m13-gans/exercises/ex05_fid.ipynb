## Concept focus — Adversarial training, mode collapse, and generative realism

GANs matter because they frame generation as a game between a generator and a critic. That game can produce sharp samples, but it is also notoriously unstable, which makes these exercises useful for building intuition about failure modes.

```text
noise -> generator -> fake sample -> discriminator
real sample ---------------> discriminator

G learns to fool D
D learns to distinguish real from fake
```

### How to think about it
View GAN training as coupled optimization, not one model training alone. If either player gets too strong or too weak, gradients stop being useful and pathologies like mode collapse appear quickly.
