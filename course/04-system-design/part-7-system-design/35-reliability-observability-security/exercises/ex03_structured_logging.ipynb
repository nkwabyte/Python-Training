### Task 1: JSON Log Formatter

Implement `format_log(level, message, correlation_id, **extra_fields)` that returns a valid JSON string containing `level`, `message`, `correlation_id`, and all extra keyword fields.


```python
# ANSWER 1
import json
import time

def format_log(level: str, message: str, correlation_id: str, **kwargs) -> str:
    payload = {
        "timestamp": time.time(),
        "level": level.upper(),
        "correlation_id": correlation_id,
        "message": message,
        **kwargs
    }
    return json.dumps(payload)

```

## Self-Check Harness


```python
def check(passed: bool, msg: str) -> bool:
    status = "PASS" if passed else "FAIL"
    print(f"{status}  {msg}")
    return passed

log_str = format_log("INFO", "Payment captured", "req_99812", amount=45.0, user_id="u_77")
parsed = json.loads(log_str)

results = [
    check(parsed["level"] == "INFO", "Task 1: Log level serialized"),
    check(parsed["correlation_id"] == "req_99812", "Task 1: Correlation ID attached"),
    check(parsed["amount"] == 45.0 and parsed["user_id"] == "u_77", "Task 1: Extra contextual fields preserved"),
]
print(f"Summary: {sum(results)}/{len(results)} checks passed.")

```
