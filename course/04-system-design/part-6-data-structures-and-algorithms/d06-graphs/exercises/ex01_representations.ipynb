## Concept focus — Graph thinking for relationships and traversal

Graphs model roads, dependencies, network hops, recommendations, workflows, and many system-design problems. Learning graph traversal is really learning how to reason about connectivity, reachability, shortest paths, and ordering constraints.

```text
A -- B -- D
|    |
C -- E

BFS: explores by layers outward from the start node
DFS: explores deeply before backing up
```

### How to think about it
Stop seeing a graph as a picture only; see it as a set of relationships. The important questions are usually: what counts as a node, what counts as an edge, and what property of the path are we optimizing or preserving?
