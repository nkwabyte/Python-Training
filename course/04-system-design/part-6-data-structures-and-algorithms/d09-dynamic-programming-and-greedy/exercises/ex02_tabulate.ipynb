## Concept focus — Optimal substructure, reuse, and local-choice strategy

Dynamic programming and greedy methods help when brute force explodes. The hard part is recognizing structure: whether subproblems repeat, whether local choices stay globally safe, and what state you must preserve to avoid recomputation.

```text
recursive problem
   |
repeated subproblems?
   | yes
memoize / tabulate

if local best choice always stays safe -> greedy may work
```

### How to think about it
Do not memorize DP as “use a table.” Start by identifying the state, the recurrence, and the decision boundary. For greedy problems, demand a reason the local rule preserves global optimality instead of trusting intuition.
