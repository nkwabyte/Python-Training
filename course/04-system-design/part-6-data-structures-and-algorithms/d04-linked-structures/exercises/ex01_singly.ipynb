## Concept focus — Pointer-linked structures and locality trade-offs

Linked structures help when inserts, deletes, and re-linking matter more than random access. They are also a bridge to queues, stacks, and caches, where structure and mutation patterns matter more than raw indexing speed.

```text
[node A] -> [node B] -> [node C] -> None

cheap relink:
A.next = C

costly access:
must walk from head to find later nodes
```

### How to think about it
Picture every node as a tiny object carrying both data and a route to the next step. Linked structures trade cache-friendly layout for flexible structural edits, which is why they appear in algorithms even when Python lists dominate day-to-day code.
