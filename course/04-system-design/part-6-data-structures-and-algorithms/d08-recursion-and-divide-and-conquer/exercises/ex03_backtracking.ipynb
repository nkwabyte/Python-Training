## Concept focus — Recursive decomposition and call-tree reasoning

Recursion is valuable because it mirrors the structure of many problems directly: trees, search spaces, and divide-and-conquer splits. The difficulty is not syntax but tracking frames, base cases, and how work accumulates across the call tree.

```text
solve(problem)
 ├─ solve(smaller part)
 ├─ solve(smaller part)
 └─ combine answers

base case stops the expansion
```

### How to think about it
Read recursive code in terms of promises. Assume the smaller call works, then ask what the current frame contributes. When recursion becomes hard to follow, draw the call tree or the stack frames explicitly.
