## Concept focus — Hierarchies, ordering, and priority management

Trees and heaps let you exploit structure instead of scanning everything. They matter because many system behaviors — scheduling, searching, indexing, balancing, and dependency ordering — become easier when you arrange data hierarchically.

```text
        8
      /        3     10
    / \         1   6      14

heap idea: parent priority <= children priority
```

### How to think about it
Think of a tree as a decision path and a heap as a partial ordering. Trees help you narrow the search space, while heaps help you repeatedly retrieve the next-best candidate without fully sorting everything.
