# Text and strings

Strings are sequences of characters. Python gives you many tools to inspect, combine, slice, and transform text. Strings are central to user input, file content, and data processing.

```text
text = "python"
indexes: 0 1 2 3 4 5
         p y t h o n

text[0] -> "p"
text[1:4] -> "yth"
```

### How to think about it
Text processing is often a matter of understanding character positions and patterns. Learning what slicing and methods do gives you a strong foundation for parsing real data.

````python
from __future__ import annotations


def make_table(rows: list[tuple[str, int]]) -> str:
    lines = [f"{name:<8} {score:>3}" for name, score in rows]
    return "\n".join(lines)
````
