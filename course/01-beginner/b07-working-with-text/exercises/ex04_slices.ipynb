# Working with text

Strings are how Python stores text. Every message, username, file path, sentence, and command-line argument starts life as a string. Because text appears everywhere, string skills become useful very quickly.

Unlike some other values, strings are **immutable**, which means their contents cannot be changed in place. Instead, string methods return new strings. That is why operations like `.strip()`, `.replace()`, or slicing produce fresh values rather than editing the old one.

This module teaches you how to clean input, split text apart, join pieces back together, and format output so it is readable. Those are practical skills you will use in tiny scripts and large applications alike.

## Task

Use slicing to extract the middle and reverse a word.


````python
from __future__ import annotations


def middle_and_reverse(word: str) -> tuple[str, str]:
    middle = word[1:-1]
    return middle, word[::-1]


if __name__ == "__main__":
    print(middle_and_reverse("python"))
````
