# Organizing code with modules and packages

As programs grow, it is better to split code into modules and packages that group related functionality. This keeps projects maintainable and makes it easier to reuse code across files.

```text
project/
├── main.py
├── helpers.py
├── math_tools.py
└── utils/
    └── format.py

main imports helpers and uses their functions
```

### How to think about it
Modular design reduces clutter. Instead of putting everything in one file, group related functions together. Imports let one file access another file’s purpose without rewriting code.
