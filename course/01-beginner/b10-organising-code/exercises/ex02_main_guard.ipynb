# Organising code into modules

As programs grow, one file quickly becomes hard to manage. Organising code means splitting related ideas into separate files and modules so that each part has a clear purpose. This makes your code easier to read, test, and reuse.

A module is just a Python file that can be imported by another Python file. Importing lets one part of your program reuse functions, classes, or constants from another part. That is a major step from writing isolated beginner scripts toward building real projects.

This module is really about structure: deciding what belongs together, what should stay separate, and how to make files work together without becoming confusing.
