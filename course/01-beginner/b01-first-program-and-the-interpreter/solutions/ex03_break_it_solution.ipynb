## Task 1: the four predictions

| Line | Error | Why |
|---|---|---|
| A `print("A: " + 5)` | `TypeError` | Text plus number. Python will not guess between joining and adding. |
| B `print(totall)` | `NameError` | A name Python has never seen. A typo for `total`. |
| C `print("C: ", 8 / (4 - 4))` | `ZeroDivisionError` | `4 - 4` is `0`, and division by zero has no answer. |
| D `print("D: unclosed` | `SyntaxError` | The string is never closed. |

**The one that catches people is C.** The zero is not written anywhere. You have
to evaluate `4 - 4` in your head to see it. Real division-by-zero bugs look
exactly like this: the zero arrives from a calculation, an empty list, or a
count that happened to be nothing today.

**And D behaves differently from the rest.** A `SyntaxError` stops the whole cell
before *any* of it runs, so in the original cell lines A, B, and C never
executed either. That is why you were asked to run them one at a time.

## Task 2: fix all four

````python
print("A: " + str(5))          # convert the number to text first
print("A: ", 5)                # or let print handle both, which is simpler

total = 42
print(total)                   # define the name before using it

divisor = 4 - 2
print("C: ", 8 / divisor)      # make sure the divisor cannot be zero

print("D: closed properly")
````

**Why two fixes for A.** `str(5)` turns the number into text so `+` has two
pieces of text to join. `print("A: ", 5)` sidesteps the problem entirely by
letting `print` take two values. The second is what you would actually write.

There is a general principle hiding here: when a fix requires a conversion, ask
first whether you needed to combine the values at all. Often you did not.

**On C.** Naming the divisor is not decoration. `8 / divisor` invites the
question "can `divisor` be zero", and `8 / (4 - 4)` hides it. Names are where
you put the thinking.

## Task 3: your own four errors

Ada's, for the shape of a good record:

````python
my_syntax_error = "SyntaxError: '(' was never closed"
my_name_error = "NameError: name 'shoping_list' is not defined"
my_type_error = "TypeError: unsupported operand type(s) for -: 'str' and 'int'"
my_zero_division_error = "ZeroDivisionError: integer division or modulo by zero"
````

**Why recording the exact wording matters.** You are building a lookup table in
your head between a sentence Python says and a thing you did. That mapping is
most of what "getting good at debugging" means, and it forms from repetition of
the exact words rather than from a paraphrase.

Notice Ada's `TypeError` mentions `-` rather than `+`, and `ZeroDivisionError`
says "integer division or modulo" because it came from `//` or `%`. Python's
messages are specific. Read them as specific.

## Task 4: the shopping list bug

````python
print("Shopping list")
print("-" * 20)
print("1. bread")
print("2. apples")          # was: "2. " + apples
print("3. coffee")
print("-" * 20)
print("Items: " + str(3))
````

**The bug was `apples` without quotes**, on line 4, giving `NameError: name
'apples' is not defined`. The word looks like text because it reads like text,
but Python saw a name.

**What made this one harder than tasks 1 and 3.** The bug is surrounded by
correct code, including a line further down that does the same kind of thing
*correctly*, `"Items: " + str(3)`. Real bugs live in working code. That is what
makes them hard, not the bug itself.

**And the thing worth noticing about the fix.** Once quoted, the `+` is
unnecessary:

```python
print("2. apples")
```

is better than

```python
print("2. " + "apples")
```

When a fix leaves behind machinery that is no longer doing anything, remove it.
Leftover scaffolding is how code becomes hard to read.
