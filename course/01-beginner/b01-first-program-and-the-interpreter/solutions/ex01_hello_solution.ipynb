## Task 1: print your name

````python
print("Ada Mensah")
````

**Why this way.** There is nothing else to it, and that is the point of a first
task. The quotes make the text into text, the brackets make `print` happen.

**Near misses worth recognising.**

| What people write | What happens |
|---|---|
| `print(Ada Mensah)` | `SyntaxError`. Without quotes Python reads `Ada` as a name it should know. |
| `print("Ada Mensah")` with smart quotes | `SyntaxError`. Word processors replace `"` with `"` and `"`. Use a code editor. |
| `Print("Ada Mensah")` | `NameError`. Python is case sensitive; `Print` is not `print`. |

## Task 2: three lines

````python
print("Ada Mensah")
print("Accra")
print("A tool that tracks my reading")
````

**Why this way.** Three separate `print` calls, because each `print` ends with a
new line. The line breaks come from `print`, not from anything you typed.

**The near miss that teaches the most:**

```python
print("Ada Mensah", "Accra", "A tool that tracks my reading")
```

That is one line with spaces between the items, not three lines. It is a
perfectly good instruction that answers a different question. If you wrote it,
you have learned what the comma does, which is worth the detour.

## Task 3: fix the three broken lines

````python
# Line 1 was: print("Half a quote)
print("Half a quote")

# Line 2 was: print Hello
print("Hello")

# Line 3 was: Print("Wrong capital letter")
print("Wrong capital letter")
````

**The predictions, and why.**

**Line 1, `SyntaxError`.** The string was opened and never closed, so Python read
to the end of the line still looking for the closing quote. Note that Python
often reports an unclosed quote or bracket on the *following* line, because that
is where it finally gave up. This is the single most confusing thing about
`SyntaxError` and it is worth remembering now.

**Line 2, `SyntaxError`.** Two mistakes at once: no brackets and no quotes.
Without brackets, `print Hello` is not an instruction Python can parse at all.

**Line 3, `NameError`.** This one is different in kind. The shape of the line is
valid Python, so it parses; Python then looks for something called `Print` and
does not find one. Shape errors are `SyntaxError` and happen before anything
runs. Missing names are `NameError` and happen during the run.

If you predicted `SyntaxError` for line 3, that is a good wrong answer. The
difference between "not shaped like Python" and "shaped fine, but I do not know
that word" is exactly what exercise 3 goes on to drill.

## Task 4: the profile block

````python
print("Name: Ada Mensah")
print()
print("Learning Python since: 12 August 2026")
````

**Why this way.** `print()` with nothing between the brackets prints nothing and
then moves to the next line, which is a blank line. That is the whole trick.

**Near misses.**

```python
print("Name: Ada Mensah\n")
```

This also produces a blank line, using `\n`, which means "new line" inside a
string. It works, and you will meet it properly in module B07. At this point in
the course `print()` says what you mean more plainly, and plainness is worth
more than cleverness for a long time yet.

```python
print("")
```

Also works. `print()` and `print("")` do the same thing here. `print()` is the
more common style because there is no empty value to explain.
