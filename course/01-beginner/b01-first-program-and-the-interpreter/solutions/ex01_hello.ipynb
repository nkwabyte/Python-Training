# Your first Python program

A Python program is just a sequence of instructions the interpreter reads and executes one line at a time. The interpreter turns your text into actions such as printing output, storing values, and evaluating expressions.

```text
source code  -->  Python interpreter  -->  computer actions
     (your file)                 (reads and runs)

print("Hello")   ->   output on screen
```

### How to think about it
Think of the interpreter as a translator that speaks both your language and the machine language the computer understands. The goal is not to memorize syntax only; it is to understand the order of execution.

Solution: your first Python program.


````python


from __future__ import annotations


def greet() -> str:
    lines = [
        "Hello, Python!",
        "This is my first script.",
        "I am learning to code.",
    ]
    return "\n".join(lines)


if __name__ == "__main__":
    print(greet())
````
