## Task 2: the three division operators

````python
what_slash_does = "Divides and always gives a decimal, even when it divides evenly. 7 / 2 is 3.5, and 4 / 2 is 2.0 rather than 2."
what_double_slash_does = "Divides and throws away the fractional part, rounding toward negative. 7 // 2 is 3."
what_percent_does = "Gives the remainder left over after the division. 7 % 2 is 1."

print(what_slash_does)
print(what_double_slash_does)
print(what_percent_does)
````

**The detail most people miss.** `4 / 2` is `2.0`, not `2`. A single slash always
produces a decimal number, whatever the inputs. That looks pedantic today and it
explains a whole class of confusion in module B02, when you find that `2` and
`2.0` are not the same kind of thing.

**Why `%` matters more than it looks.** The remainder answers "is this divisible
by that", which is how you check for even numbers, wrap a value around a range,
and do anything on a cycle. `n % 2 == 0` is one of the most common lines in
beginner Python.

**The near miss.** Describing `//` as "rounds down" is nearly right and fails on
negatives: `-7 // 2` is `-4`, not `-3`. It rounds toward negative rather than
toward zero. Not something to memorise now, but if you wrote "rounds down",
that is where it stops being true.

## Task 4: the five predictions

````python
print("5" + "5")   # 55     joins two pieces of text
print(5 + 5)       # 10     adds two numbers
print("5" * 3)     # 555    repeats the text three times
print(5 * 3)       # 15     multiplies
# print("5" + 5)   # TypeError
````

**The whole lesson of module B02, arriving early.**

`+` means join for text and add for numbers. `*` means repeat for text and
multiply for numbers. The operator does not decide; the *kinds of value* on
either side decide.

And the last line is the one that matters:

```python
"5" + 5
```

`TypeError: can only concatenate str (not "int") to str`.

Python could have guessed. It could have decided you meant `"55"`, or `10`. It
refuses, because both guesses are defensible and a language that guesses wrong
silently is a language that produces wrong answers without telling you.

That refusal is a design decision you will come to be grateful for. Languages
that do guess here are a well known source of bugs that survive to production.

**If you got fewer than two predictions wrong**, either you have programmed
before or you were being cautious. The task was designed to catch people out.
