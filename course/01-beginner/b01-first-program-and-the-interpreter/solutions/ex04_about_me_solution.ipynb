````python
print("=" * 40)
print("  Ada Mensah")
print("=" * 40)
print()
print("Location:      Accra, Ghana")
print("Learning:      Python")
print("Started:       12 August 2026")
print("Why:           To build tools I actually use")
print()
print("=" * 40)
````

## Task 2: the same card as one string

````python
print("""========================================
  Ada Mensah
========================================

Location:      Accra, Ghana
Learning:      Python
Started:       12 August 2026
Why:           To build tools I actually use

========================================""")
````

**Which would I rather maintain? The multi-print version.**

That is a considered answer rather than an obvious one, and here is the
reasoning.

The triple-quoted version is *what the output looks like*, which makes it easy to
read and easy to eyeball. But the border is now a literal row of forty equals
signs typed by hand. Change the width and you retype it, twice, and count. In
the first version it is `"=" * 40`, and changing the width is changing one
number in two places.

The triple-quoted version also silently loses the ability to put anything
computed into it. The moment one line needs today's date rather than a typed
date, that block has to come apart.

**Where the judgement flips.** If the text is genuinely fixed, a help message, a
banner, a licence notice, then the triple-quoted version is better, because it
shows the shape and nothing about it will ever be calculated.

The rule underneath: **choose the form that matches what the text actually is.**
Fixed text, one block. Assembled text, separate lines. If your answer differed
but you gave a reason of this kind, your answer is fine.

## Task 3: `sep` and `end`

````python
print("coffee", "long walks", "clean tracebacks", sep=", ")
print(1, 2, 3, 4, 5, sep=" -> ")
````

**Why `sep` rather than building the string yourself.** You could write

```python
print("coffee" + ", " + "long walks" + ", " + "clean tracebacks")
```

and get the same output. It is longer, it repeats the separator twice, and
changing the separator means three edits instead of one. `sep` says what you mean
once.

**The near miss.**

```python
print("coffee, ", "long walks, ", "clean tracebacks", sep="")
```

This puts the separator inside the values and then removes the automatic one. It
produces the right output by cancelling one mistake with another. If you wrote
it, notice that the commas are now part of the data rather than part of the
formatting, and that is the kind of tangle that becomes a real problem when the
data comes from somewhere else.

## Task 4: make it better

````python
WIDTH = 44

print("+" + "-" * (WIDTH - 2) + "+")
print("|" + "Ada Mensah".center(WIDTH - 2) + "|")
print("+" + "-" * (WIDTH - 2) + "+")
print()
print("  Location   Accra, Ghana")
print("  Learning   Python, from nothing")
print("  Started    12 August 2026")
print("  Why        To build tools I actually use")
print()
print("+" + "-" * (WIDTH - 2) + "+")
print("  b01 complete".ljust(WIDTH))
````

**What changed and why it reads better.**

The width is named once, as `WIDTH`, and every line is computed from it. Change
one number and the whole card resizes. In task 1 the width was typed four times
and the alignment was counted by hand.

The border became a box with corners rather than three loose rules, which makes
the card look like one object instead of three separated regions.

`center` and `ljust` do the spacing arithmetic instead of a human doing it. You
have not been taught these; they came from `help(str)` in exercise 2, which is
the point of having taught you `help`.

The labels lost their colons and gained a fixed column, which reads more like a
table and less like a list of sentences.

**What to take from this.** None of it is about `print`. It is about noticing
that a repeated value wants a name, that a computed layout survives editing and
a typed one does not, and that "looks deliberate" comes from consistency rather
than from decoration. Those judgements carry into every part of this course.

**If you did not use `center` or `ljust`, nothing is wrong.** They were not
taught. Getting there with `"=" * 40` and counted spaces is a complete answer to
the task as set.
