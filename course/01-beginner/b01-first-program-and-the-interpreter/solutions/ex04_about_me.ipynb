Solution: print a profile block.


```python


from __future__ import annotations


def profile_card(name: str, city: str, age: int, hobby: str) -> str:
    lines = [
        f"Name: {name}",
        f"City: {city}",
        f"Age: {age}",
        f"Favourite hobby: {hobby}",
    ]
    return "\n".join(lines)


if __name__ == "__main__":
    print(profile_card("Ada", "London", 28, "reading"))
```
