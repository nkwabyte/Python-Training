## 2. What a program actually is

A program is a list of instructions, written as plain text, that a piece of
software called an **interpreter** reads one line at a time and carries out.

For Python the interpreter is a program called `python`. It reads your line,
works out what you asked for, does it, then moves to the next line. There is no
magic step in between, and nothing happens that you did not write.

That is the entire model. Everything else in this course is detail on top of it.

---

## 3. Your first instruction: `print`

Here is a complete Python program. One line.

Read it first, then run it.

````python
print("Hello, world!")
````

It printed `Hello, world!` underneath. Now take the line apart, because every
piece of it matters.

```
print ( "Hello, world!" )
  |    |        |       |
  |    |        |       └── closing bracket, ends the instruction
  |    |        └────────── the thing you want shown, in quotes
  |    └─────────────────── opening bracket, starts the list of things to show
  └──────────────────────── the name of the instruction
```

**`print`** is the name of a piece of built-in behaviour. Python already knows
what it means. You did not have to define it.

**The brackets** `(` and `)` are how you say "do it, and here is what to do it
with". The name on its own does nothing; the brackets are what make it happen.

**The quotes** `"` and `"` mark where your text starts and ends. Everything
between them is treated as text rather than as instructions. Without them Python
would try to make sense of the words, and it cannot.

**What `print` does:** it shows a value where you can see it. That is all. It
does not store anything, change anything, or send anything anywhere. It is how
your program talks to you, and for the next few modules it is your only window
into what your code is doing.

## 4. Prove it to yourself

The fastest way to understand a rule is to watch it fail. Run each cell below
and read what comes back before moving on.

First, the name with no brackets.

````python
print
````

Python told you what `print` is, a built-in function, but it did not print
anything. You named the instruction without asking for it to happen.

Now the text without quotes.

````python
print(Hello, world!)
````

That is an **error**, and it is the first one you have seen. Look at the last
line of the red block. It says something like:

```
SyntaxError: invalid syntax
```

**Syntax** means the shape of the line, the grammar. Python is telling you the
line is not shaped like Python. Without the quotes it tried to read `Hello` as
an instruction it should already know, and it does not know one by that name.

**The rule that matters more than any other in this course: read the error from
the bottom.** The last line tells you what went wrong. The lines above it tell
you where. Beginners read from the top, see a wall of unfamiliar words, and give
up before reaching the one line that was written for them.

Errors are not failures. They are the interpreter telling you precisely what it
could not do. You will see thousands of them and they will stop feeling like
anything.

## 5. Single or double quotes

Both work, and they mean exactly the same thing. Run this.

````python
print('Hello, world!')
print("Hello, world!")
````

Identical output. Use whichever you like, but stay consistent inside a project.

The reason both exist is that it lets you put one kind of quote inside the
other, which you will need the moment you write an apostrophe.

````python
print("It's a nice day")
````

Try changing that line to use single quotes on the outside, run it, and read
the error. Then change it back. Doing this now costs you fifteen seconds and
saves you a confused afternoon later.

## 6. Printing more than one thing

Give `print` several values, separated by commas, and it shows them all with a
space between each.

````python
print("Hello,", "world!")
print("one", "two", "three")
````

Notice you did not have to add the spaces yourself. `print` puts one between
each item by default. You can change that, but not yet; for now just know that
the space came from `print` and not from your text.

Each `print` also starts a new line. Two prints, two lines. That is how you will
produce several lines of output.

## 7. Comments

there for the human reading the code, which is usually you in three weeks having
forgotten everything.

````python
# This line does nothing at all.
print("But this line runs.")  # A comment can also sit after code.
````
