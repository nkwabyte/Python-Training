# Your first Python program

This notebook is about the most basic idea in programming: giving the computer a clear instruction and seeing what happens. In Python, a program is a set of lines written in a file or notebook cell. When you run the cell, Python reads those lines and executes them in order. That is the starting point for everything else you will learn.

One of the first useful instructions is `print()`. The job of `print()` is simple: show a value on the screen. It helps beginners because it gives immediate feedback. If you are unsure what your code is doing, printing values is often the fastest way to make the program visible.

At this stage, do not worry about advanced theory. Focus on the cycle of reading code, running it, looking at the output, and changing something small. That loop is how programming becomes familiar.

## This exercise: hello world and first output

Run the code cell, observe what appears underneath it, then change the message and run it again. The goal is to connect written code with visible output.

Create a script that prints exactly three lines of output.


```python


from __future__ import annotations


def greet() -> str:
    lines = [
        "Hello, Python!",
        "This is my first script.",
        "I am learning to code.",
    ]
    return "\n".join(lines)


if __name__ == "__main__":
    print(greet())
```
