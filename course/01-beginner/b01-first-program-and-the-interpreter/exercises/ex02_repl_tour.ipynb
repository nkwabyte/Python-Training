## Exercise 2: The REPL, Python's laboratory

In exercise 1 you ran Python inside a notebook. There is a second place Python
lives, and it is the one you will reach for most often once you are working: a
prompt in a terminal where Python answers you one line at a time.

It is called the **REPL**, which stands for Read, Evaluate, Print, Loop. Python
**reads** what you type, **evaluates** it, **prints** the answer, then **loops**
back and waits for the next thing. That is literally all it does.

## 1. Why bother, when notebooks exist

A notebook is for work you are keeping. The REPL is for questions you are asking.

When you cannot remember whether `len` works on a number, the REPL answers in
four seconds and forgets you asked. You do not create a file, you do not name
anything, you do not save. It is the difference between a laboratory bench and
a lab report.

Experienced Python programmers keep a REPL open all day. It is the single
fastest way to answer a question about the language, faster than a search
engine, and it is never out of date because it is your actual Python.

## 3. The one difference that confuses everybody

In a file, printing something requires `print`. In the REPL, it usually does not.

Type each of these at the `>>>` prompt, pressing Enter after each.

```
2 + 2
"hello"
print("hello")
```

You will see:

```
>>> 2 + 2
4
>>> "hello"
'hello'
>>> print("hello")
hello
```

Three things to notice, and they matter.

**`2 + 2` printed `4` without you asking.** The REPL automatically shows the
value of whatever you type. That is the P in REPL. A file does not do this. Put
`2 + 2` alone in a file and nothing at all appears.

**`"hello"` came back with quotes around it, as `'hello'`.** The REPL is showing
you the value *as Python would write it*, so you can tell a piece of text apart
from a number. That is a different job from displaying it to a user.

**`print("hello")` came back with no quotes.** `print` shows the text itself,
the way you would want a person to read it.

So the REPL echoes values for you, the programmer, and `print` displays values
for the person using your program. The quotes are the visible difference.

---

## 4. The two keys that save you the most time

**The up arrow** brings back the last thing you typed. Press it again for the one
before that. You almost never retype a line in the REPL; you bring it back and
edit it. This is how you try six variations of an idea in a minute.

**`exit()`** closes the REPL and returns you to your ordinary terminal. Typing
`exit` on its own without brackets gets you a polite message reminding you about
the brackets, which is Python being unusually kind.

Try both now. Run something, press up to bring it back, change one character,
press Enter. Then leave with `exit()` and open a new REPL, because you will
notice the next thing.

---

## 5. The REPL forgets everything

Open a fresh REPL and type:

```
>>> 2 + 2
4
```

Now `exit()` and open it again. The `4` is gone, along with everything else you
did. Nothing you type in a REPL survives closing it.

This is the reason files exist. **The REPL is for finding out. A file is for
keeping.** When something you discovered in the REPL turns out to be worth
having tomorrow, that is the moment to move it into a `.py` file.

Beginners often try to write whole programs at the `>>>` prompt and then lose
them. Now you know why not to.

---

## 6. Three things to try while you are in there

Type these and read what comes back. You are not expected to understand the
answers yet. You are learning that the REPL will answer questions.

```
>>> type(4)
>>> type("4")
>>> len("hello")
```

`type` tells you what kind of thing a value is. Notice that `4` and `"4"` are
not the same kind of thing, which will matter enormously in module B02.

`len` tells you how long something is.

And this one, which is the most useful habit in this whole course:

```
>>> help(print)
```

That is the real documentation for `print`, built into Python, available
offline, always correct for your version. Press `q` to leave it.
