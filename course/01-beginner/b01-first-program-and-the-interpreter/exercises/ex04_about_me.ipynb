## 1. An empty print makes a blank line

`print` with nothing between the brackets prints nothing, then moves to the next
line. That gives you a blank line, which is the cheapest way to give output room
to breathe.

```python
print("first")
print()
print("third")
```

---

## 2. Text can be multiplied

This looks strange the first time and it is genuinely useful.

```python
print("-" * 30)
print("=" * 30)
print("ab" * 5)
```

`"-" * 30` means thirty copies of that character joined together. It is how
every separator line you have ever seen in a terminal is made.

Notice this is the same `*` that multiplies numbers. Python decides what `*`
means from the kinds of value on either side. Text times a number repeats it;
number times number multiplies. And as you found in exercise 3, text plus number
is refused outright, because there is no sensible guess to make.

---

## 3. `sep` and `end` change the joins

When you give `print` several values it puts a space between them and a new line
at the end. Both are settings you can change.

```python
print("a", "b", "c")
print("a", "b", "c", sep="")
print("a", "b", "c", sep=" | ")
```

```python
print("no newline after this ", end="")
print("so this continues the same line")
```

`sep` is what goes *between* the values. `end` is what goes *after* the last one,
and it is a new line unless you say otherwise.

You do not need to understand the `name=value` form yet. That is a keyword
argument and module B06 explains it properly. For now: `sep` and `end` are two
dials on `print`, and they have sensible defaults.

---

## 4. Triple quotes hold several lines

Three quote marks open a block of text that can run over as many lines as you
like, keeping the line breaks exactly as you typed them.

```python
print("""Name:    Ada
Role:    Engineer
Status:  Learning Python""")
```

This is the readable choice when the text is genuinely a block. Several separate
`print` calls are the readable choice when the lines are separate thoughts. Pick
by which one the text actually is.
