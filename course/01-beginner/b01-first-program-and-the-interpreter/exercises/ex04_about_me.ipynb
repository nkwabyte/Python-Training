# Strings, quotes, and printed output

Programs often need to display information in a form a person can read. That could be a greeting, a profile, a receipt, or a status message. Early on, this is where strings become useful. A string is Python’s type for text.

When you build a short profile block, you practice several beginner skills at once: storing values, placing text next to values, and formatting output so it is easy to read. That may seem small, but readable output is part of writing readable programs.

This exercise also helps you get comfortable with quotes and f-strings. Those are basic tools you will use constantly throughout the course.

## This exercise: print a small profile card

Run the example, then change the name, city, age, and hobby so the output becomes your own.

## Task

Build a simple profile card for a learner and print it in a neat format.


```python


from __future__ import annotations


def profile_card(name: str, city: str, age: int, hobby: str) -> str:
    lines = [
        f"Name: {name}",
        f"City: {city}",
        f"Age: {age}",
        f"Favourite hobby: {hobby}",
    ]
    return "\n".join(lines)


if __name__ == "__main__":
    print(profile_card("Ada", "London", 28, "reading"))
```
