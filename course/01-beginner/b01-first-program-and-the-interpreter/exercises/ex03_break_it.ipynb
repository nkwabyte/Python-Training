## 1. An error is a message, not a punishment

When Python cannot do what you asked, it stops and tells you exactly where it
got to and what it could not do. It is not scolding you. It has no opinion about
you at all. It is reporting.

The report is called a **traceback**, because it traces back through the steps
that led to the problem.

Run the cell below. It will produce an error. That is intended.

````python
print("this line is fine")
print(undefined_thing)
print("this line never runs")
````

## 2. Reading a traceback

Look at what came back. Ignore the volume of it and find these three things.

```
Traceback (most recent call last):          <- a heading, ignore it
  Cell In[1], line 2                        <- WHERE: line 2 of this cell
    print(undefined_thing)                  <- the actual line
NameError: name 'undefined_thing' is not defined
^^^^^^^^^                                   ^^^^^^^^^
WHAT went wrong                             the detail
```

**Read from the bottom.** The last line is the one written for you. It names the
kind of error, then explains it in a sentence.

The lines above it tell you where. When your program grows, there will be many
of them, forming a chain from where you started to where it broke. The bottom
of that chain is usually the part you wrote.

Notice also that `"this line is fine"` printed, and `"this line never runs"` did
not. Python executes top to bottom and stops dead at the error. Anything after
it never happened. That fact alone will explain a great many confusing moments.

**`SyntaxError`** means the line is not shaped like Python, so Python could not
even begin. Here the opening bracket was never closed.

A syntax error is special: it happens *before anything runs*. That is why you
sometimes see one reported on a line that looks perfectly fine, when the real
problem is an unclosed bracket or quote on the line above. **When a SyntaxError
points at a line you are sure is correct, look at the line before it.**

````python
message = "hello"
print(mesage)
````

**`NameError`** means you used a name Python has never heard of. Ninety percent
of the time it is a typo, as it is here: `mesage` against `message`. The rest of
the time it is a name you meant to create but did not, or created inside
something else.

````python
print("You are " + 30 + " years old")
````

**`TypeError`** means the values were the wrong kind for the operation. You
cannot add a number to text with `+`, because `+` means join for text and add
for numbers, and Python refuses to guess which you meant.

This is the error that teaches the lesson at the centre of module B02: `30` and
`"30"` are different kinds of thing.

````python
print(10 / 0)
````

**`ZeroDivisionError`** means exactly what it says. Not every error is subtle.
Some are just arithmetic reminding you that it has rules.

## 4. The procedure

When something breaks, in this course or in ten years, do this in this order.

1. **Read the last line.** It names the error and explains it.
2. **Find the line number.** It tells you where to look.
3. **Read the line above the failing one.** Especially for `SyntaxError`.
4. **Say the error out loud in your own words.** If you cannot, you have not
   understood it yet, and changing code before you understand it is guessing.
5. **Change one thing.** Then run again. Changing three things means you will
   not know which one helped.

That is it. There is no step six, and there is no version of this procedure that
gets more sophisticated later.
