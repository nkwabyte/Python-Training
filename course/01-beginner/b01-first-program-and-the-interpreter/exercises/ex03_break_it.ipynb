# Your first Python program

A Python program is just a sequence of instructions the interpreter reads and executes one line at a time. The interpreter turns your text into actions such as printing output, storing values, and evaluating expressions.

```text
source code  -->  Python interpreter  -->  computer actions
     (your file)                 (reads and runs)

print("Hello")   ->   output on screen
```

### How to think about it
Think of the interpreter as a translator that speaks both your language and the machine language the computer understands. The goal is not to memorize syntax only; it is to understand the order of execution.

# Making mistakes on purpose

A beginner’s first error message can feel intimidating, but it should not. Errors are one of the normal ways Python communicates with you. They tell you that the interpreter could not continue and give clues about what went wrong.

Different kinds of mistakes produce different exceptions. A misspelled name may cause a `NameError`. A broken string or bracket may cause a `SyntaxError`. Using the wrong type of value in an operation may produce a `TypeError`. Learning these patterns early makes later debugging much easier.

This exercise is valuable because it makes failure safe. Instead of waiting for a confusing bug later, you will trigger common mistakes on purpose and inspect what Python says about them.
