## Task

Build a small command loop for add and summary actions.


````python
from __future__ import annotations


def command_loop(command: str, expenses: list[dict[str, object]]) -> float:
    if command == 'sum':
        return sum(float(item['amount']) for item in expenses)
    return 0.0


if __name__ == "__main__":
    print(command_loop('sum', [{"amount": 3.0}, {"amount": 4.0}]))
````
