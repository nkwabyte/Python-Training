## Task

Reject negative amounts and blank descriptions.


```python
from __future__ import annotations


def valid_expense(description: str, amount: float) -> bool:
    return bool(description.strip()) and amount > 0


if __name__ == "__main__":
    print(valid_expense('coffee', 4.5))
    print(valid_expense('', -1))
```
