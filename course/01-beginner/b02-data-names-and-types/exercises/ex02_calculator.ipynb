## 1. Two kinds of number

`int` is a whole number. `float` is a number with a decimal point.

```python
whole = 7
decimal = 7.5

print(whole, type(whole))
print(decimal, type(decimal))
```

`type(x)` tells you what kind of thing `x` is. `<class 'int'>` and
`<class 'float'>` are the two you will use constantly.

Python's `int` has no maximum. This is unusual and worth seeing.

```python
print(2 ** 200)
```

In most languages that overflows. In Python it does not, because `int` grows to
fit. Floats do have limits, and they have something stranger, which is section 3.

## 2. The operators

Five you will use every day, and two that are worth knowing now.

```python
print("7 + 2 =", 7 + 2)
print("7 - 2 =", 7 - 2)
print("7 * 2 =", 7 * 2)
print("7 / 2 =", 7 / 2)      # true division: always a float
print("7 // 2 =", 7 // 2)    # floor division: drops the fraction
print("7 % 2 =", 7 % 2)      # remainder
print("7 ** 2 =", 7 ** 2)    # power
```

Two details that catch people.

**`/` always gives a float**, even when it divides exactly. `4 / 2` is `2.0`, not
`2`. If you need a whole number, `//` is the operator you meant.

**`%` gives the remainder**, and it answers "is this divisible by that". `n % 2`
is `0` for even numbers and `1` for odd. You will use it more than you expect.

```python
print(4 / 2, type(4 / 2))
print(4 // 2, type(4 // 2))
print("is 10 even?", 10 % 2 == 0)
```

## 3. Floats are approximate, and it shows

Run this. It is not a bug in Python.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

`0.30000000000000004`, and the comparison is `False`.

A float stores a number in binary, and 0.1 has no exact binary representation,
in the same way that one third has no exact decimal representation. You write
0.333... and stop somewhere; the computer does the same thing in base two.

**What to do about it, today:**

- Never compare floats with `==`. Compare with a tolerance, or work in whole
  units.
- For money, do not use floats at all. Work in the smallest unit, so 1999 pence
  rather than 19.99 pounds, or use `Decimal`, which intermediate module 03 covers.

This is not a Python flaw. Every language using standard floating point behaves
this way, and most beginners in every one of them discover it by having a total
come out one penny wrong.

```python
total = 0.0
for _ in range(10):
    total += 0.1

print(total)
print(total == 1.0)
print(abs(total - 1.0) < 1e-9)     # the comparison you should have written
```

## 4. `input` gives you text, always

`input()` reads a line typed by the user and hands it back **as text**, whatever
it looks like.

Notebooks block waiting for typing, which is awkward mid-lesson, so this notebook
uses a stand-in that behaves the same way. In a real script you would call
`input()` directly.

```python
def pretend_input(prompt, typed):
    """Behaves like input(), but you supply what was typed."""
    print(prompt + typed)
    return typed          # a str, exactly like the real input()


answer = pretend_input("How many apples? ", "5")
print("value:", answer)
print("type: ", type(answer))
```

`'5'` with quotes, and `<class 'str'>`. It is text that looks like a number.

Now watch what that does to arithmetic.

```python
first = pretend_input("First number: ", "5")
second = pretend_input("Second number: ", "3")

print("sum:", first + second)
```

`53`, not `8`.

`+` between two pieces of text means join, so `"5" + "3"` is `"53"`. Nothing has
gone wrong; Python did exactly what you asked. You asked the wrong thing.

Every programmer meets this once. You have now met it somewhere it cost nothing.

## 5. Converting

`int()` and `float()` turn text into a number. `str()` goes the other way.

```python
text = "5"
number = int(text)

print(text, type(text))
print(number, type(number))
print("now the arithmetic works:", number + 3)
```

Conversion can fail, and it fails loudly rather than guessing.

```python
int("banana")
```

`ValueError: invalid literal for int() with base 10: 'banana'`.

**`ValueError` means the type was right but the value was not.** You gave `int`
a string, which is what it wants, but that particular string is not a number.
Compare it with `TypeError`, which means the type itself was wrong.

One more that surprises people:

```python
int("3.5")
```

Also a `ValueError`. `int` parses whole numbers only. `"3.5"` is a valid float,
not a valid int, and Python will not silently discard the `.5` for you.

The fix is to say what you meant:

```python
print(float("3.5"))          # parse it as a float
print(int(float("3.5")))     # parse, then deliberately truncate
```

## What you learned

- `int` is a whole number and has no maximum. `float` has a decimal point.
- `/` always produces a float. `//` drops the fraction. `%` gives the remainder
  and answers divisibility.
- Floats are approximate. `0.1 + 0.2` is not `0.3`, never compare them with `==`,
  and never store money in one.
- `input()` always returns text, even when the text looks like a number.
- `"5" + "3"` is `"53"`, because `+` joins text.
- `int()`, `float()`, and `str()` convert on purpose. Failure is a `ValueError`,
  which means the value was wrong, as against `TypeError`, which means the type
  was wrong.
