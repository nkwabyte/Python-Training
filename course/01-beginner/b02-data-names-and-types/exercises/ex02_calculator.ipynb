# Variables, names, and types

A variable is a label attached to a value stored in memory. The name helps you refer to data later, while the type tells you what kind of data it is: text, whole number, decimal, boolean, and so on.

```text
name: age
      |
      v
value: 18
Type: int

name: name
      |
      v
value: "Ada"
Type: str
```

### How to think about it
Use meaningful names like `total_cost` instead of `x`. A variable is not just a box; it is a way to track changing information in the program.

# Data, names, and basic Python types

In Python, a **name** is a label attached to a value. When you write `x = 10`, Python binds the name `x` to the value `10`. If you later write `x = 99`, Python does not edit the old value; it simply points `x` at a new value. This idea matters because it helps you reason clearly about assignment, copying, and bugs caused by reusing names.

This module also introduces the types beginners use most often: `int` for whole numbers, `float` for decimal numbers, `str` for text, and `bool` for true/false answers. A big part of early Python is learning to ask: **what type is this value right now, and what type do I need before I use it?**

You should also remember that `input()` always gives you text. Even if the user types `42`, Python receives `"42"`, not the number `42`. That is why conversion functions like `int()` and `float()` exist.

As you work through these exercises, pay attention to names, values, and types. Those three ideas appear in every Python program you will ever write.

## Task

Read two values, convert them to numbers, and print the four arithmetic results.


````python
from __future__ import annotations


def calculate(left_text: str, right_text: str) -> dict[str, float]:
    left = float(left_text)
    right = float(right_text)
    return {
        "sum": left + right,
        "difference": left - right,
        "product": left * right,
        "quotient": left / right,
    }


if __name__ == "__main__":
    values = calculate(input("First number: "), input("Second number: "))
    for key, value in values.items():
        print(f"{key}: {value}")
````
