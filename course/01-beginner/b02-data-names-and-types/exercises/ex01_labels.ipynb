## 1. Assignment attaches a label

Run this.

````python
score = 10
print(score)
````

The `=` is called **assignment**. Read it right to left: make the value `10`,
then attach the name `score` to it.

```
score = 10

  score ──────> 10
    |            |
  the label   the value
```

The name is not the value. It points at it. That distinction does nothing at all
until section 4, and then it does everything.

---

## 2. Rebinding moves the label

A name can be pointed at something else later. Nothing about the old value
changes; the label just moves.

````python
score = 10
print("before:", score)

score = 25
print("after: ", score)
````

The `10` was not modified. The name `score` stopped pointing at `10` and started
pointing at `25`.

That sounds like hair-splitting when the values are numbers. Hold on to it.

---

## 3. Two names can point at the same value

This is where the box model starts to fail.

````python
a = 10
b = a          # b now points at the same value a points at
print(a, b)

a = 99         # this moves the label a, and does not touch b
print(a, b)
````

`b` was set from `a`, then `a` was rebound. `b` still shows `10`.

With the box model you would say "b got a copy of the value", and you would
predict the same output. So far both models agree. The next section is where they
disagree, and where the box model gets people hurt.

---

## 4. The case the box model gets wrong

Some values can be **changed in place**. A list is one of them. You will meet
lists properly in module B05; all you need now is that `[1, 2, 3]` is a list and
`.append(4)` adds to it.

**Predict what the last line prints before you run this.**

````python
a = [1, 2, 3]
b = a
b.append(4)

print("b is", b)
print("a is", a)
````

If you predicted `[1, 2, 3]` for `a`, you were using the box model, and you have
just watched it fail. This is the single most common surprise in Python.

Here is what actually happened.

```text
a = [1, 2, 3]        a ──────> [1, 2, 3]

b = a                a ──┐
                         ├───> [1, 2, 3]
                     b ──┘

b.append(4)          a ──┐
                         ├───> [1, 2, 3, 4]     one object, changed
                     b ──┘
```

There was never a second list. `b = a` attached a second label to the same
object. `b.append(4)` changed that object, and `a` points at it, so `a` shows
the change.

**Mutation changes the object. Rebinding changes the name.** Two operations,
similar syntax, opposite consequences. Almost every confusing bug involving
shared data is somebody confusing those two.

---

## 5. Proving it with `id`

`id(x)` gives you a number identifying which object `x` points at. If two names
have the same `id`, they point at the same object.

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]      # a different list that happens to look the same

print("a and b same object?", id(a) == id(b))
print("a and c same object?", id(a) == id(c))
print("a and c equal?      ", a == c)
````

`a` and `c` are **equal**, because they contain the same things. They are not
**identical**, because they are two separate objects.

Equal is about contents. Identical is about which object. Python spells the
second one `is`.

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)    # same object
print(a is c)    # different objects that happen to be equal
print(a == c)    # but their contents match
````

---

## 6. Rebinding does not affect the other label

````python
a = [1, 2, 3]
b = a
b = [9, 9, 9]      # REBIND b: point it at a brand new list

print("a is", a)   # unchanged, because a still points at the old one
print("b is", b)
````

Compare this cell with section 4. The only difference is `b = [9, 9, 9]` instead
of `b.append(4)`, and the consequence is opposite.

| Operation | What it does | Does the other name see it? |
|---|---|---|
| `b.append(4)` | changes the object | yes |
| `b = [9, 9, 9]` | moves the label `b` | no |

## 7. Names that say what they hold

A name is documentation that the computer checks the spelling of. Rules:

- `snake_case`, lowercase with underscores. Not `camelCase`.
- Say what it holds, not what type it is. `customer_email`, not `str2`.
- Length in proportion to how far apart the name is defined and used.
- Never name something after a built-in. `list = [1, 2]` breaks `list()` for the
  rest of the session, and the error appears far away from the cause.

Watch that last one happen, so you recognise it later.

````python
list = [1, 2, 3]     # a terrible idea, done here on purpose
print(list)
````

````python
# now try to use the real list() function
list("abc")
````

`TypeError: 'list' object is not callable`.

Read the last line. It says the thing you tried to call is a list object, not a
function, because the name `list` no longer refers to the built-in. The traceback
points at the line that failed, not at the line that broke it, which is exactly
what makes this bug annoying.

Restart the kernel to get `list` back.

## What you learned

- Assignment attaches a name to a value. The name is a label, not a box.
- Rebinding moves a label and leaves the object alone.
- Two names can point at one object, and then a change through one is visible
  through the other.
- **Mutation changes the object; rebinding changes the name.** Confusing those
  two is the most common source of surprising bugs in Python.
- `==` compares contents. `is` asks whether it is the same object. `id()` proves
  it.
- Naming something after a built-in breaks that built-in, and the error surfaces
  far from the cause.
