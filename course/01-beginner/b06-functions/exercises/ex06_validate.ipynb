# Functions and reusable code

A **function** is a named block of code that performs one job. Functions help you avoid repetition, make programs easier to test, and give useful names to pieces of logic. Instead of rewriting the same steps many times, you place them in one function and call it when needed.

Good beginner functions usually do one clear thing, take in the values they need as parameters, and return a result when they are finished. A return value is different from printing. `print()` shows something to a person, while `return` gives a value back to the rest of the program.

As you work through this module, focus on three questions: what inputs does this function need, what output should it produce, and what name would make its purpose obvious to another learner reading the code?

```python
from __future__ import annotations


def validate_age(age: int) -> str:
    if age < 0:
        return "negative age"
    if age > 120:
        return "age is too high"
    return "valid"


if __name__ == "__main__":
    for age in [-1, 25, 200]:
        print(age, validate_age(age))
```
