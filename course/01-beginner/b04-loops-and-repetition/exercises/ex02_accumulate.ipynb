A loop that prints is a loop that forgets. Each pass throws away what it saw and
starts the next one empty handed.

This notebook teaches the pattern that gives a loop a memory. It has three parts
and they live in three different places, and every one of the classic loop bugs
is one of those three parts put in the wrong place. You will run two of those
bugs on purpose.

## 2. The accumulator pattern

To carry something from one pass to the next, you need a name that lives outside
the loop. That name is called an **accumulator**.

The pattern has three parts, and where each one sits is the whole lesson.

```
total = 0                 <- 1. START IT EMPTY, before the loop, at no indent
│
for reading in readings:
    total = total + reading   <- 2. ADD TO IT, inside the loop, indented
│   └────────────────────┘
│
print(total)              <- 3. USE IT, after the loop, back at no indent
```

Start empty. Add in the loop. Use after. Say it until it is boring, because the
three failures at the end of this notebook are each one of those three lines put
somewhere else.

````python
readings = [12.4, 15.1, 9.8, 21.3, 17.0, 8.2]

total = 0
for reading in readings:
    total = total + reading

print("total:", total)
````

Follow the value of `total` through the passes. This table is what the loop
does, written out:

| Pass | `reading` | `total` before | `total` after |
|---|---|---|---|
| start | | | 0 |
| 1 | 12.4 | 0 | 12.4 |
| 2 | 15.1 | 12.4 | 27.5 |
| 3 | 9.8 | 27.5 | 37.3 |
| 4 | 21.3 | 37.3 | 58.6 |
| 5 | 17.0 | 58.6 | 75.6 |
| 6 | 8.2 | 75.6 | 83.8 |

Writing that table by hand for a loop you do not understand is the single most
effective debugging technique in this module. It takes two minutes and it is
never wrong.

---

## 3. `+=` is shorthand for the same thing

`total = total + reading` is written so often that Python has a short form.

````python
total = 0
total = total + 5
print("long form: ", total)

total = 0
total += 5
print("short form:", total)
````

`total += reading` and `total = total + reading` do the same job. The short form
is what you will see in other people's code, so this notebook uses it from here
on.

`-=`, `*=`, and `/=` exist too and work the same way.

---

## 4. Counting is the same pattern with a different step

To count, add 1 instead of adding the value. Put an `if` inside the loop, from
module B03, and you are counting only the items you care about.

````python
readings = [12.4, 15.1, 9.8, 21.3, 17.0, 8.2]

warm_count = 0
for reading in readings:
    if reading > 15:
        warm_count += 1

print("readings above 15:", warm_count)
````

Three parts again, in the same three places. Only the middle line changed.

Note where the `if` sits: indented once, inside the loop. And `warm_count += 1`
sits indented twice, inside the `if`, inside the loop. Each level of indentation
answers a question about when that line runs.

---

## 5. Collecting is the same pattern with a list

An accumulator does not have to be a number. Start with an **empty list**,
written `[]`, and add to it with `.append(x)`, which puts `x` on the end.

````python
readings = [12.4, 15.1, 9.8, 21.3, 17.0, 8.2]

warm = []
for reading in readings:
    if reading > 15:
        warm.append(reading)

print("the warm ones:", warm)
print("how many:    ", len(warm))
````

Same three parts. `warm = []` starts it empty, `warm.append(reading)` adds
inside the loop, `print` uses it after.

This is the pattern that replaces the whole of the wrong instinct beginners have
about changing a list while walking through it. Exercise 5 shows what happens to
people who follow that instinct instead.

---

## 6. The average needs both a total and a count

`len` gives you the count without a loop, so an average is a division after the
loop.

````python
readings = [12.4, 15.1, 9.8, 21.3, 17.0, 8.2]

total = 0
for reading in readings:
    total += reading

average = total / len(readings)
print("average:", average)
print("average, rounded:", round(average, 2))
````

`round(x, 2)` gives you `x` to two decimal places. It is for display. Do not
round a value you are going to keep calculating with, because the error grows
with every step.

## 7. The reset in the wrong place, run on purpose

Now a real bug. The code below is one keystroke away from the loop in section 2:
the `total = 0` line has been indented by four spaces, which puts it inside the
loop.

Predict what it prints before you run it.

````python
readings = [12.4, 15.1, 9.8, 21.3, 17.0, 8.2]

for reading in readings:
    total = 0
    total += reading

print("total:", total)
````

`8.2`, which is the last reading.

No traceback. No warning. A number came out, it is the right type, it has a
plausible size, and it is wrong.

Here is what happened. Part one of the pattern moved inside the loop, so it ran
once per pass. Every pass threw away everything the loop had accumulated and
started again from zero. The value that survived is whatever the final pass
added.

**The symptom is worth memorising: a total that equals the last item means the
accumulator is being reset inside the loop.** Recognising that symptom turns a
two-hour hunt into a ten-second fix, and you will meet it more than once.

---

## 8. The same bug, with a traceback instead

Now the same shape over an empty list.

The accumulator is called `tally` here rather than `total`, and that is
deliberate. `total` already exists in this notebook from the cells above, and a
leftover value from an earlier cell would answer for it and hide the error. A
name that has never been used gives you the honest result.

That is worth knowing in its own right: a notebook remembers everything you have
run, so a cell that works can depend on a name you deleted ten cells ago.
Restarting the kernel is how you find out.

````python
readings = []

for reading in readings:
    tally = 0
    tally += reading

print("total:", tally)
````

```
NameError: name 'tally' is not defined
```

**`NameError`** means you used a name that does not exist yet.

The list was empty, so the body never ran, so the line that creates `tally`
never ran, so by the time `print` asked for it there was nothing to give. This
is the same defect as section 7. The data decided which symptom you got.

That is the real lesson. One bug, two faces: a silently wrong number when there
is data, a crash when there is not. Starting the accumulator above the loop
fixes both at once, and it is why part one of the pattern lives where it lives.

````python
readings = []

tally = 0                     # part one, above the loop, where it belongs
for reading in readings:
    tally += reading

print("total:", tally)        # 0, which is the honest answer for no readings
````

## 9. One more failure: the average of nothing

`total` is now safe. The division is not.

````python
readings = []

total = 0
for reading in readings:
    total += reading

average = total / len(readings)
print(average)
````

```
ZeroDivisionError: division by zero
```

**`ZeroDivisionError`** says what it says. `len(readings)` is 0, and nothing
divided by zero has an answer.

An empty list is falsy, which you met in module B03, so the guard is short:

````python
readings = []

total = 0
for reading in readings:
    total += reading

if readings:
    print("average:", total / len(readings))
else:
    print("no readings, so no average")
````

Deciding what to print when there is no data is a design decision, not an
accident. `0` would be a lie: it says the average temperature was zero degrees.

## What you learned

- A list holds items in order, `len` says how many, and a `for` loop walks
  through it.
- The accumulator pattern has three parts: **start it empty above the loop, add
  to it inside the loop, use it after the loop.**
- `total += x` is the short form of `total = total + x`.
- Counting is the same pattern adding 1, usually inside an `if`.
- Collecting is the same pattern with `[]` and `.append`.
- An average needs the total and `len`, and rounding is for display only.
- An accumulator reset inside the loop gives a total equal to the last item, and
  no error at all. That symptom names the bug.
- The same misplaced line raises `NameError` when the data is empty, and
  dividing by `len` of an empty list raises `ZeroDivisionError`.
