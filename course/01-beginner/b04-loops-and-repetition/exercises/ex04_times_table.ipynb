## Exercise 4: A loop inside a loop

Anything laid out in a grid needs two loops: one for the rows and one for the
things in a row. Tables, calendars, board games, seating plans, and every report
with columns are all this shape.

Nesting loops is not a new rule. It is the rule you already have, applied twice.
What is new is the arithmetic: two nested loops do not add their work together,
they multiply it, and that multiplication is worth noticing before it surprises
you.

By the end you will have printed a twelve times table that lines up.

## 1. The inner loop runs all the way through, every pass of the outer

```
for row in [1, 2, 3]:              <- the outer loop, three passes
│
    for column in ["a", "b"]:      <- the inner loop, two passes each time
    │
        print(row, column)         <- the body, indented twice
```

Read it as a sentence: for each row, walk the whole of the columns.

````python
for row in [1, 2, 3]:
    for column in ["a", "b"]:
        print(row, column)
````

Six lines. The outer name changes slowly, the inner name changes quickly.

```
row = 1   column = a
row = 1   column = b     <- inner finished, outer moves on
row = 2   column = a
row = 2   column = b
row = 3   column = a
row = 3   column = b
```

The inner loop starts again from its first item every time. It does not carry on
from where it left off, and nothing remembers that it has run before.

---

## 2. Work does not add, it multiplies

Count the passes rather than assuming.

````python
passes = 0

for row in range(1, 4):
    for column in range(1, 5):
        passes += 1

print("outer passes: 3")
print("inner passes each time: 4")
print("body ran:", passes, "times")
````

Three and four give twelve, not seven.

That is fine at this size. Here is why it is worth naming anyway. When both
loops walk the same collection of `n` things, the body runs `n * n` times.

| Items | Body runs |
|---|---|
| 10 | 100 |
| 100 | 10,000 |
| 1,000 | 1,000,000 |
| 10,000 | 100,000,000 |

Ten times more data, a hundred times more work. A report that returned instantly
on your test file of 50 rows can take minutes on the real file of 5,000, and
nothing about the code will look different.

This is not a reason to avoid nested loops. Grids need them. It is a reason to
know when you have written one over a large collection, and to be able to say
how large. The intermediate and advanced levels put proper numbers on this and
give it its name.

## 3. `print` starts a new line unless you tell it otherwise

Every `print` you have written finished with a newline. That is a setting with a
default, not a law.

`print(x, end="")` finishes with nothing, so the next `print` carries on along
the same line. The default is `end="\n"`, where `\n` means newline.

````python
for n in range(1, 6):
    print(n, end="")

print()          # an empty print supplies the newline that was skipped
print("that ran on one line")
````

The bare `print()` at the end matters. Without it the next thing printed would
continue on the same line, and in a notebook the output of two cells would run
together.

In a grid, that empty `print()` is what ends a row. Where you put it decides
whether you get a grid or a column, and section 6 shows what happens when it is
in the wrong place.

## 4. Making columns line up

Numbers have different widths, so a grid printed without padding is a mess. Two
text operations fix it.

`"7".rjust(4)` pads text on the left with spaces until it is 4 characters wide,
which pushes the content to the right. `"-" * 20` repeats text 20 times.

````python
print("[" + "7".rjust(4) + "]")
print("[" + "42".rjust(4) + "]")
print("[" + "144".rjust(4) + "]")
print("-" * 20)
````

Every one of those is four characters wide, so a column of them lines up.

Both are **text** operations. A number has to be turned into text first, with
`str()`, which module B02 covered.

````python
n = 144
print("[" + str(n).rjust(4) + "]")
````

## 5. Forgetting `str`, run on purpose

````python
n = 144
print(n.rjust(4))
````

```
AttributeError: 'int' object has no attribute 'rjust'
```

**`AttributeError`** means you asked a value for something it does not have. The
part after the dot is called an attribute, and `rjust` is one that text has and
whole numbers do not.

This error type is the one that tells you the value is not the kind of thing you
thought it was. Read it as a question: what type did I actually have here. The
answer is in the message, and here it is `'int'`.

The fix is `str(n).rjust(4)`: turn it into text, then pad the text.

## 6. One row at a time

Now the pieces go together. The inner loop prints the cells of one row with
`end=""`, and the empty `print()` ends the row.

Look hard at the indentation of that `print()`. It is indented once, so it
belongs to the outer loop and runs once per row.

````python
size = 5

for row in range(1, size + 1):
    for column in range(1, size + 1):
        print(str(row * column).rjust(4), end="")
    print()
````

Now the same code with that one line indented one level further, which puts it
inside the inner loop. Run it.

````python
size = 5

for row in range(1, size + 1):
    for column in range(1, size + 1):
        print(str(row * column).rjust(4), end="")
        print()
````

Twenty-five lines instead of five, one number on each.

No traceback. No warning. Every number is correct and every one of them is in
the wrong place. The row ending now happens after every cell rather than after
every row, because that is what its indentation says.

This is the nested-loop version of the bug you met in exercise 2, where a line
indented one level too far ran once per pass instead of once. **In a nested
loop, the indentation of a line is the entire statement of when it runs.** There
is no other clue and no error to help you.

When a grid comes out as a column, look at that line first.

## 7. A header, and the finished table

A times table wants labels. The header is a separate loop before the grid,
because it runs once rather than once per row.

````python
size = 5

print("    ", end="")                       # the empty corner
for column in range(1, size + 1):
    print(str(column).rjust(4), end="")
print()

print("    " + "-" * (4 * size))

for row in range(1, size + 1):
    print(str(row).rjust(4), end="")        # the row label
    for column in range(1, size + 1):
        print(str(row * column).rjust(4), end="")
    print()
````

Four separate pieces, each doing one job: the header row, the rule under it, the
row label, and the cells. Changing `size` changes all four at once, because none
of them has a `5` written into it.

A width of 4 holds numbers up to 999, so this table survives up to a size of 31.
Widths chosen by guessing break silently when the data grows, and the symptom is
a grid whose columns stop lining up.

## What you learned

- The inner loop runs all the way through on every pass of the outer loop, and
  starts again from its first item each time.
- Nested loops multiply their work. Two loops over the same `n` items run the
  body `n * n` times, so ten times the data is a hundred times the work.
- `print(x, end="")` leaves the cursor on the line, and a bare `print()` supplies
  the newline that was skipped.
- `str(n).rjust(4)` pads to a fixed width so columns line up, and `"-" * 20`
  repeats text.
- `n.rjust(4)` on a whole number is an `AttributeError`, which is the error type
  that means the value is not the kind of thing you thought.
- In a nested loop the indentation of a line is the whole statement of when it
  runs. A row ending in the wrong place turns a grid into a column, silently.
- A header is a separate loop, because it runs once rather than once per row.
