## 1. `while` repeats for as long as a condition holds

Here is the shape, taken apart.

```
while attempts < 3:
│     └────┬────┘└── the colon ends the header line and promises a block
│          │
│          └── a condition: anything that produces True or False
└── the keyword

    print("asking")
└──┘
  │
  └── four spaces. The body, run again and again while the condition holds.
```

The condition is checked **before every pass**, including the first. If it is
`False` at the start, the body never runs at all.

Compare the two loops:

| | `for` | `while` |
|---|---|---|
| Number of passes | fixed when the loop starts | decided as it goes |
| Needs a counter you manage | no | usually yes |
| Can run forever | not over a `range` | yes |

Reach for `for` when you know the items. Reach for `while` when the stopping
point depends on something that happens inside the loop.

````python
attempts = 0

while attempts < 3:
    print("attempt", attempts)
    attempts += 1

print("finished after", attempts, "attempts")
````

---

## 2. The three parts of a correct `while`

That loop stopped because three separate things were true at once. All three
have to be there, and each lives somewhere different.

```
attempts = 0              <- 1. SET UP the variable, before the loop
│
while attempts < 3:       <- 2. TEST it in the condition
│     └──────┬─────┘
│            │
    print("attempt", attempts)
    attempts += 1         <- 3. CHANGE it in the body, towards making the
                                test false

print("finished")
```

Set up. Test. Change towards false. Miss any one of the three and the loop is
broken, and it is broken in a way that costs you differently each time.

- Miss part one and you get a `NameError` at once. That is the kind failure.
- Miss part two and you have written `while True`, which is sometimes what you
  meant and needs a `break` to escape.
- Miss part three, or change in the wrong direction, and the loop never ends.
  Exercise 6 of this module is entirely about that case and shows it safely.

---

## 3. Missing part one, run on purpose

The counter below is called `tries` rather than `attempts`, because `attempts`
already exists from the cell above and a leftover value would answer the
condition and hide the error. A notebook remembers everything you have run, so
a name that has never been used is the only honest way to demonstrate this one.

````python
while tries < 3:
    print("attempt", tries)
    tries += 1
````

```
NameError: name 'tries' is not defined
```

**`NameError`** means the name does not exist. The condition is the first thing
Python evaluates, before the body ever runs, so it asked for `tries` and there
was nothing to give.

The body would have created it. The body never ran. A condition cannot be
answered with a value the loop was going to make later.

This is the same lesson as the accumulator in exercise 2. The variable a loop
depends on is set up **above** the loop.

## 4. `break` leaves the loop immediately

`break` stops the loop where it stands. The rest of the body is skipped and the
condition is not checked again.

````python
n = 0

while n < 10:
    print("n is", n)
    if n == 3:
        print("  found what I wanted, leaving")
        break
    n += 1

print("after the loop, n is", n)
````

The condition said `n < 10` and the loop left at 3. `break` overrules the
condition.

`break` works in a `for` loop too, and means the same thing there.

````python
for name in ["ama", "kofi", "yaa", "kwame"]:
    print("checking", name)
    if name == "yaa":
        print("  that is the one")
        break
````

## 5. `continue` skips the rest of this pass

`continue` abandons the current pass and goes back to the top for the next one.
Nothing below it in the body runs for that pass.

````python
for n in [4, -1, 7, -3, 2]:
    if n < 0:
        print(n, "is negative, skipping")
        continue
    print(n, "doubled is", n * 2)
````

The negative values never reach the doubling line.

`continue` in a `while` loop carries a trap that `continue` in a `for` loop does
not. In a `while`, the line that changes your variable is part of the body, and
`continue` skips the rest of the body. If the change sits below the `continue`,
it stops happening and the loop never ends.

The habit that avoids it: in a `while` with a `continue` anywhere in it, put the
change **at the top of the body**, above every branch. Every loop in this
notebook does that. Exercise 6 shows you the version that does not.

## 6. Input, without freezing the notebook

`input()` reads a line typed by the user and hands it back as text. It also
stops everything until somebody types, which makes a notebook useless as a
teaching tool.

So this notebook uses a stand-in with the same behaviour, where you supply what
was typed. In a real script you would call `input()` directly and delete the
scripted answers.

````python
def pretend_input(prompt, typed):
    """Behaves like input(), but you supply what was typed."""
    print(prompt + typed)
    return typed          # a str, exactly like the real input()


answer = pretend_input("How many? ", "5")
print("value:", answer, "  type:", type(answer))
````

A loop needs a different scripted answer each pass, so the answers go in a list
and the loop takes one at a time by **position**.

`scripted[0]` is the first item, `scripted[1]` the second, and counting starts
at zero, which matches `range` starting at zero. Exercise 5 covers positions
properly. Here you need only that much.

````python
scripted = ["ten", "-4", "7"]

print(scripted[0])
print(scripted[1])
print(scripted[2])
print("how many scripted answers:", len(scripted))
````

## 7. Text that looks like a number, and text that does not

Everything from `input` is text, which module B02 established. Converting it can
fail.

````python
typed = pretend_input("How many? ", "seven")
print(int(typed))
````

```
ValueError: invalid literal for int() with base 10: 'seven'
```

**`ValueError`** means the type was right and the value was not. `int` wanted
text, got text, and that particular text is not a number.

You cannot stop a user typing `seven`. What you can do is check before
converting. `.isdigit()` asks a piece of text whether every character in it is a
digit.

````python
print('"42".isdigit()    ->', "42".isdigit())
print('"seven".isdigit() ->', "seven".isdigit())
print('"-3".isdigit()    ->', "-3".isdigit())
print('"3.5".isdigit()   ->', "3.5".isdigit())
print('"".isdigit()      ->', "".isdigit())
````

Read those last three rows carefully, because they are where `.isdigit()` will
let you down.

- `"-3"` is `False`. The minus sign is not a digit.
- `"3.5"` is `False`. The dot is not a digit either.
- `""` is `False`, which is the answer you want for empty input.

So `.isdigit()` means "is a run of digits", not "is a number". For whole
positive numbers, which is what a guessing game wants, it is exactly right. For
anything that can be negative or fractional it is the wrong check, and the
intermediate level covers the one you want instead.

Say plainly what a validator accepts. A validator whose limits you have not
named is a bug waiting for a user with a minus sign.

## 8. A validation loop

Now put it together. Ask until the answer is usable, with `continue` to reject
and `break` to accept.

````python
scripted = ["ten", "", "-4", "7"]

index = 0
value = None

while index < len(scripted):
    typed = pretend_input("How many? ", scripted[index])
    index += 1                      # the change, at the top, above every branch

    if not typed.isdigit():
        print("  that is not a whole number, asking again")
        continue

    value = int(typed)
    print("  accepted:", value)
    break

print("final value:", value)
````

Three rejections, then an acceptance.

Two details are load bearing. `index += 1` sits at the top of the body, so the
`continue` cannot skip it. And the condition is `index < len(scripted)`, so the
loop cannot outrun its scripted answers even if none of them is ever accepted.

With a real `input()` the condition would be `while True:` and the `break` would
be the only way out, because a real user can keep typing. The scripted version
needs the bound.

## 9. A retry limit, and knowing how you left

A retry limit is a counter tested in the condition. The awkward part is what
comes after: the loop can end two ways, and from outside you cannot see which.

The answer is a **flag**, an ordinary `bool` set inside the loop and read after
it.

````python
secret = 42
scripted = ["fifty", "80", "40", "42"]

attempts = 0
max_attempts = 4
won = False                          # the flag, set up before the loop

while attempts < max_attempts:
    typed = pretend_input("Your guess: ", scripted[attempts])
    attempts += 1

    if not typed.isdigit():
        print("  whole numbers only")
        continue

    guess = int(typed)

    if guess == secret:
        print("  correct, in", attempts, "attempts")
        won = True
        break
    elif guess < secret:
        print("  too low")
    else:
        print("  too high")

if won:
    print("well played")
else:
    print("out of attempts, the number was", secret)
````

`won` is the accumulator pattern from exercise 2 with a `bool` instead of a
number: set up above the loop, changed inside it, read after it.

Without the flag you would have to guess from `attempts`, and `attempts` is 4
both when the fourth guess was right and when it was wrong. A value that cannot
tell two outcomes apart is not an answer.

## What you learned

- `while` repeats for as long as its condition holds, and the condition is
  checked before every pass including the first.
- A correct `while` has three parts: set the variable up above the loop, test it
  in the condition, and change it in the body towards making the test false.
- Missing the set-up gives a `NameError` before the body has run once.
- `break` leaves the loop at once and overrules the condition.
- `continue` abandons the current pass. In a `while`, put the line that changes
  your variable at the top of the body so `continue` cannot skip it.
- `input()` blocks a notebook, so a scripted stand-in stands in for it, and a
  list of scripted answers is read by position.
- `int()` on text that is not a number raises `ValueError`. `.isdigit()` checks
  first, and it says `False` for a minus sign, a decimal point, and empty text.
- A loop that can end two ways needs a flag set inside it and read after it.
