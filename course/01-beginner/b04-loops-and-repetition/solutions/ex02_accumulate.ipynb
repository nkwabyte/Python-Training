```python
from __future__ import annotations


def analyse(readings: list[int], threshold: int) -> tuple[int, int, list[int]]:
    total = 0
    above_threshold = 0
    kept: list[int] = []
    for value in readings:
        total += value
        if value > threshold:
            above_threshold += 1
            kept.append(value)
    return total, above_threshold, kept
```
