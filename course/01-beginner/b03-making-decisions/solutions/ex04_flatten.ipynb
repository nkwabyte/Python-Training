```python
from __future__ import annotations


def ticket_status(age: int, has_ticket: bool, with_parent: bool) -> str:
    if age >= 18 and has_ticket:
        return "allowed"
    if age >= 18 and not has_ticket:
        return "ticket required"
    if with_parent and has_ticket:
        return "allowed"
    if with_parent and not has_ticket:
        return "ticket required"
    return "too young"
```
