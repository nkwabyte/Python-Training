# If statements and decision-making

An if statement checks a condition and decides whether to run a block of code. Conditions evaluate to either True or False, and those booleans control the flow of the program.

```text
if score >= 70:
    print("Pass")
else:
    print("Retry")

Condition -> True  ?  run pass block
           False ?  run else block
```

### How to think about it
Every decision is a fork in the road. The computer checks one condition at a time, and only the matching branch runs. Try to read conditions as questions: “Is this true?”

````python
from __future__ import annotations


def is_valid_password(password: str) -> bool:
    return (
        len(password) >= 8
        and any(ch.isdigit() for ch in password)
        and any(ch.isalpha() for ch in password)
        and " " not in password
    )
````
