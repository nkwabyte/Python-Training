## 1. A comparison is a question, and its answer is a value

`>` does not print anything and does not change anything. It produces a value:
`True` or `False`. You met `bool` in module B02; this is where those two values
start earning their keep.

````python
print("7 > 3   is", 7 > 3)
print("3 > 7   is", 3 > 7)
print("7 == 7  is", 7 == 7)
print("7 != 7  is", 7 != 7)
print("7 >= 7  is", 7 >= 7)
print("7 <= 3  is", 7 <= 3)

print("the type of a comparison result:", type(7 > 3))
````

Six operators, and you now have all of them.

| Operator | Asks |
|---|---|
| `==` | are these equal |
| `!=` | are these different |
| `>` | is the left one greater |
| `<` | is the left one smaller |
| `>=` | greater or equal |
| `<=` | smaller or equal |

Two of those need separating, because they look alike and do opposite jobs.

- `=` attaches a name to a value. It is an instruction.
- `==` asks whether two values are equal. It is a question.

`score = 90` changes your program. `score == 90` asks about it and changes
nothing. Section 8 shows what happens when the two get swapped.

Comparisons work on text as well as numbers. `"a" < "b"` is `True`, in
dictionary order. `"10" < "9"` is also `True`, because text is compared
character by character and `"1"` comes before `"9"`. That is one more reason to
convert user input to a number before comparing it.

````python
print("abc" == "abc")
print("a" < "b")
print("10" < "9")        # text order, not number order
print(10 < 9)            # number order
````

## 2. `if` runs a block when the answer is `True`

Here is the shape, taken apart.

```
if score >= 50:
│  └────┬────┘└── the colon ends the header line and promises a block
│       │
│       └── the condition: anything that produces True or False
│
└── the keyword

    print("pass")
└──┘
  │
  └── four spaces. This line belongs to the if.

print("done")
│
└── no spaces. This line is outside the if and always runs.
```

````python
score = 72

if score >= 50:
    print("pass")

print("done")
````

Now change one thing. The block does not run, and Python does not complain about
it. A condition that is `False` produces silence, not an error.

````python
score = 20

if score >= 50:
    print("pass")

print("done")
````

Only `done` printed. Nothing went wrong. Remember that silence: it is what a
skipped block looks like, and it is also what a wrong condition looks like.

## 3. Indentation is syntax, not decoration

In many languages, brackets mark where a block starts and stops, and the
indentation is a courtesy to human readers. In Python the indentation **is** the
block. Changing it changes the meaning of the program.

The convention, followed by essentially all Python code, is four spaces per
level.

Now a failure, run on purpose.

A `SyntaxError` is found before any of a cell runs, so a cell containing one
cannot run at all, not even its first line. To show you one anyway, the broken
code below is kept in a piece of text and handed to `exec`, which runs Python
source stored as text. You will not need `exec` in your own work. It is here so
that the traceback can be printed and read.

````python
broken = '''if 10 > 5:
print("no indentation at all")
'''

exec(broken)
````

Read it from the bottom.

```
IndentationError: expected an indented block after 'if' statement on line 1
```

**`IndentationError`** is the error type. The cause is named in full: line 1
ended with a colon, which promises a block, and line 2 did not deliver one.

Above that, `File "<string>", line 2` tells you where. `<string>` means the
source came from text handed to `exec` rather than from a file, which is only
true because of how this demonstration is built. In your own work that line
gives you a filename.

The fix is to indent the body by four spaces.

## 4. The second indentation error: tabs mixed with spaces

A tab character and a run of spaces can look identical on screen and be
completely different to Python. This is the one indentation problem you cannot
see.

In the cell below, the second line is indented with a single tab character and
the third with eight spaces. Look at them. You cannot tell.

````python
mixed = '''if 10 > 5:
	print("this line begins with one tab")
        print("this line begins with eight spaces")
'''

exec(mixed)
````

```
TabError: inconsistent use of tabs and spaces in indentation
```

**`TabError`** is a kind of `IndentationError`. The cause is that Python cannot
decide whether those two lines are at the same depth, because a tab is not a
fixed number of spaces.

Hunting this by eye costs an afternoon, because the two lines look the same in
every editor you will try. Set your editor once to insert four spaces when you
press the tab key, and you never meet this error again. Every editor has that
setting.

## 5. `else` covers everything the condition missed

`if` on its own does one thing or nothing. `if` with `else` does one thing or
the other, and exactly one of the two blocks runs, always.

````python
score = 41

if score >= 50:
    print("pass")
else:
    print("fail")
````

`else` takes no condition, because its condition is fixed: everything the `if`
did not catch. That is why it is written `else:` and not `else something:`.

---

## 6. `elif` is not a second `if`

For a third answer you need a third branch. Python spells it `elif`, short for
"else if".

Here are two versions of the same intention. Run both.

````python
score = 95

print("with separate ifs:")
if score >= 90:
    print("  A")
if score >= 80:
    print("  B")
if score >= 70:
    print("  C")
````

````python
score = 95

print("with an elif chain:")
if score >= 90:
    print("  A")
elif score >= 80:
    print("  B")
elif score >= 70:
    print("  C")
````

The first version printed three grades. The second printed one.

```
separate ifs                  an elif chain

if  ──> test, maybe run       if   ──> test, maybe run
if  ──> test, maybe run            └── if it ran, skip the rest
if  ──> test, maybe run       elif ──> only tested if nothing above ran
                              elif ──> only tested if nothing above ran
```

Three separate `if` statements are three independent questions. Every one of
them is asked, whatever the others answered. An `elif` chain is one question
with several possible answers. Testing stops at the first branch that is `True`.

A score of 95 really is greater than 90, greater than 80, and greater than 70.
Both versions did what you wrote. Only one did what you meant.

**Use `elif` when the branches are alternatives. Use separate `if` statements
when the tests are independent.**

## 7. The order of the chain decides the answer

This next cell contains no error. It runs, it prints, and it is wrong.

The score is 95. Watch what grade it gets.

````python
score = 95

if score >= 60:
    grade = "D"
elif score >= 70:
    grade = "C"
elif score >= 80:
    grade = "B"
elif score >= 90:
    grade = "A"
else:
    grade = "F"

print("score", score, "gets grade", grade)
````

`D`.

There is no traceback here, nothing in red, and nothing in the output that looks
like a complaint. That is what makes this the most expensive mistake in the
module. An error stops you; a wrong answer travels.

The cause: 95 is greater than or equal to 60, so the first branch matched and
the chain stopped. The three branches below it were never tested. They cannot be
reached by any score at all, because every score that would reach them was
already caught above.

**In a chain of thresholds on the same value, the most demanding test goes
first.** Ordered the other way, every later branch is dead code.

Two habits that catch this:

- Test the extremes, not the middle. A score of 75 gets the right grade in both
  the correct and the broken version. A score of 95 does not.
- Read your chain aloud from the top, with the word "otherwise" in front of
  every `elif`. "95 or more, otherwise 80 or more" makes sense. "60 or more,
  otherwise 70 or more" does not, and you can hear it.

````python
score = 95

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("score", score, "gets grade", grade)
````

## 8. `=` where you meant `==`

Section 1 said the two are different. Here is what Python does when they are
swapped inside a condition. This one fails loudly.

````python
confused = '''score = 90
if score = 90:
    print("top marks")
'''

exec(confused)
````

```
SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?
```

**`SyntaxError`** is the error type: the line is not valid Python at all. The
cause is that `if` needs a question, and `score = 90` is an instruction. The
caret marks the part it could not accept, and the message names the likely fix.

Python treats this as a syntax error on purpose. In several other languages the
same line is legal, quietly assigns the value, and then treats it as a
condition, which produces a bug that can survive for years. Here it cannot get
past the first run.

Ignore `:=` for now. It is a different operator with a different job, and you do
not need it in this module.

## What you learned

- A comparison produces a value, `True` or `False`. There are six operators, and
  `==` asks a question while `=` gives an instruction.
- `if` runs a block when its condition is `True`, and stays silent when it is
  not.
- Indentation is syntax. The block is defined by it, four spaces per level.
- Two indentation errors: `IndentationError` when a promised block is missing,
  and `TabError` when tabs and spaces are mixed, which you cannot see on screen.
- `else` covers everything the condition missed, and takes no condition itself.
- `elif` is one question with several answers. Separate `if` statements are
  several independent questions, and all of them get asked.
- The order of an `elif` chain decides the answer. A chain ordered from the
  loosest test downwards gives wrong results and no error at all.
- `if x = 5` is a `SyntaxError`, because `if` needs a question rather than an
  instruction.
