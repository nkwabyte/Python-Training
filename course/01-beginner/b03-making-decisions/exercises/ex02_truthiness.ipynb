## 2. `bool(x)` asks any value for its truth

`bool` converts, in exactly the way `int` and `str` converted in module B02. It
takes any value and gives back `True` or `False`.

````python
print(bool(7))
print(bool(0))
print(bool("hello"))
print(bool(""))
````

`if` does this conversion for you. These two cells mean the same thing.

````python
name = "Ada"

if bool(name):
    print("the long way")

if name:
    print("the way it is written in practice")
````

A value that converts to `True` is called **truthy**. One that converts to
`False` is called **falsy**. Those are the words the Python community uses, and
they are worth knowing because they appear in every error report and code
review you will read.

---

## 3. The falsy values, in full

The list is short, and it is worth memorising, because everything not on it is
truthy.

| Value | Why it is falsy |
|---|---|
| `False` | itself |
| `None` | Python's word for no value at all |
| `0` | zero, as an `int` |
| `0.0` | zero, as a `float` |
| `""` | the empty piece of text |
| `[]` | the empty list |
| `()` | the empty tuple |
| `{}` | the empty dictionary |

Lists arrive properly in module B05, and tuples and dictionaries after that. You
do not need them today. The pattern is the part that matters: **zero of a number
is falsy, and empty of a container is falsy.** Everything else, including `-1`,
including `"0"`, including `"False"`, is truthy.

`None` is new here. It is a value that means "nothing", and it is what a program
uses to say a thing has not been supplied yet. Write it with a capital N and no
quotes.

````python
print("None on its own prints as:", None)
print("bool(None) is", bool(None))
print("type(None) is", type(None))
````

Watch two of the traps, because they are the ones that get people.

````python
print('bool(" ")   is', bool(" "))       # a space is a character, so not empty
print('bool("0")   is', bool("0"))       # text containing a zero is not a zero
print('bool([0])   is', bool([0]))       # a list holding one falsy thing is not empty
````

`" "` contains a space, so it is not empty. `"0"` is text, and its emptiness is
what counts, not what it says. `[0]` is a list with one item in it, so the list
is not empty, whatever that item happens to be.

## 4. Why `if items` beats `if len(items) > 0`

`len(x)` gives the number of items in a container, or the number of characters
in a piece of text.

````python
print(len("Ada"))
print(len(""))
print(len([1, 2, 3]))
````

So `if len(basket) > 0:` works. It is also longer, and it is fragile in a way
that `if basket:` is not. `len` refuses anything that has no length.

The next cell fails on purpose.

````python
basket = None      # nothing was supplied

print(len(basket) > 0)
````

```
TypeError: object of type 'NoneType' has no len()
```

**`TypeError`** is the error type: the operation was fine, the type it was given
was not. `NoneType` is the type of `None`. The cause is that `len` needs
something with a length, and `None` has none.

Now the same situation, written the short way.

````python
basket = None

if basket:
    print("there is something in the basket")
else:
    print("nothing in the basket")
````

No error. `bool(None)` is `False`, so the `else` branch runs.

That is the real argument for `if basket:`. It is not only shorter. It handles
empty and missing with the same line, and it reads as a sentence: if there is a
basket, do this.

**Say what you mean, though.** `if basket:` says "if the basket has anything in
it". If you truly need to know whether the basket exists, that is a different
question, and section 7 has it.

## 5. `not` flips a truth value

`not` takes one value and gives back the opposite truth.

````python
print("not True  is", not True)
print("not False is", not False)
print('not ""    is', not "")
print('not "Ada" is', not "Ada")
print("not 0     is", not 0)
````

`not` always produces `True` or `False`, whatever it was given. `not "Ada"` is
`False`, not some flipped piece of text.

This gives you the empty check, which reads exactly as it sounds.

````python
basket = []

if not basket:
    print("your basket is empty")
````

## 6. Truthy is not the same as `== True`

This is where truthiness gets misused. `"hello"` is truthy. `"hello" == True` is
`False`. They are different questions.

````python
print('bool("hello")   is', bool("hello"))
print('"hello" == True is', "hello" == True)

print("bool(7)         is", bool(7))
print("7 == True       is", 7 == True)
print("1 == True       is", 1 == True)
print("0 == False      is", 0 == False)
````

`== True` compares against one specific value. Only `True` itself, and the
number `1`, are equal to it. Every other truthy value fails that comparison
while still being perfectly truthy.

**Never write `if x == True:`.** Write `if x:`. It is shorter, it is what you
meant, and it does not quietly exclude every truthy value that is not `1`.

The same goes for `if x == False:`. Write `if not x:`.

## 7. When truthiness is the wrong tool

Truthiness treats "empty" and "missing" as the same thing. Sometimes they are
not, and then it hides a bug.

A shop lets you order zero of something to remove it from your basket. Zero is a
real, meaningful quantity here.

````python
quantity = 0        # the customer deliberately asked for none

if quantity:
    print("order recorded:", quantity)
else:
    print("no quantity supplied")      # wrong: a quantity was supplied
````

The message is wrong. A quantity of zero was supplied, and the program reported
that nothing was.

When zero or the empty string is a legitimate value, ask the question you
actually mean. Module B02 gave you `is` for asking whether two names point at the
same object, and there is exactly one `None` in a running Python program, so
`is None` is the way to test for it.

````python
quantity = 0

if quantity is None:
    print("no quantity supplied")
else:
    print("order recorded:", quantity)

quantity = None

if quantity is None:
    print("no quantity supplied")
else:
    print("order recorded:", quantity)
````

Both cases now report correctly.

The rule to carry away: **`if x:` when empty and missing deserve the same
treatment; `if x is None:` when they do not.** Deciding which one you are in is
your job, and it is not a detail. Getting it wrong is how a system ends up
telling a customer their order of zero was never received.

## What you learned

- `if` accepts any value, not only a comparison.
- `bool(x)` converts any value to `True` or `False`, and `if` does that
  conversion for you. Truthy and falsy are the words for the two outcomes.
- The falsy values are `False`, `None`, `0`, `0.0`, `""`, `[]`, `()`, and `{}`.
  Zero of a number is falsy, empty of a container is falsy, everything else is
  truthy.
- `if items:` handles empty and missing in one line. `len` raises a `TypeError`
  when handed something with no length, such as `None`.
- `not` flips a truth value and always returns `True` or `False`.
- `if x == True:` is not the same question as `if x:`, and it silently excludes
  most truthy values. Do not write it.
- When `0` or `""` is a legitimate value, truthiness is the wrong tool and
  `is None` is the right one.
