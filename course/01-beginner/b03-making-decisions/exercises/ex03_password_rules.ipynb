Real decisions rarely rest on one fact. A password is accepted when it is long

## 1. What you are carrying in

A comparison produces `True` or `False`. `if` runs its indented block when the
condition is true, `elif` offers the next alternative and is only tested when
everything above it failed, and `else` catches the rest.

One run, so this notebook stands on its own.

````python
length = 6

if length >= 12:
    print("long")
elif length >= 8:
    print("acceptable")
else:
    print("too short")
````

One more thing carried in, because everything below depends on it: a condition
is a value, so you can give it a name.

````python
length = 6
long_enough = length >= 8

print(long_enough, type(long_enough))

if long_enough:
    print("accepted")
else:
    print("rejected")
````

`long_enough` holds `False`. Naming a condition costs one line and turns the
`if` into something a reader can follow without doing the comparison in their
head.

## 2. Four questions you can ask about a piece of text

Each of these produces `True` or `False`, so each can go straight into a
condition. Module B04 covers text properly; these four are the ones this
notebook needs.

````python
password = "Sunday42"

print("len(password)          ", len(password))
print("len(password) >= 8     ", len(password) >= 8)
print('"n" in password        ', "n" in password)
print('" " not in password    ', " " not in password)
print("password.lower()       ", password.lower())
print("password != .lower()   ", password != password.lower())
print("password.isalpha()     ", password.isalpha())
````

| Written | Asks |
|---|---|
| `len(text) >= 8` | is it at least eight characters long |
| `"n" in text` | does that appear anywhere inside it |
| `" " not in text` | is that absent from it |
| `text != text.lower()` | is there a capital in it, since lowering it changed it |
| `text.isalpha()` | is every character a letter |

`.lower()` gives back a lowercased copy and leaves the original alone. So
`text != text.lower()` is `True` exactly when lowering changed something, which
means there was a capital.

`.isalpha()` is `True` only when every character is a letter, so
`not text.isalpha()` means "contains at least one thing that is not a letter":
a digit, a symbol, or a space.

````python
print("Sunday42".isalpha())       # has digits, so False
print("Sunday".isalpha())         # all letters, so True
print("Sun day".isalpha())        # a space is not a letter, so False
````

## 3. `and` is true only when both sides are

````python
print("True  and True  is", True and True)
print("True  and False is", True and False)
print("False and True  is", False and True)
print("False and False is", False and False)
````

````python
password = "Sunday42"

long_enough = len(password) >= 8
has_capital = password != password.lower()

if long_enough and has_capital:
    print("passes both rules")
else:
    print("fails at least one rule")
````

`and` chains as far as you need. `a and b and c` is true only when all three
are.

## 4. `or` is true when either side is

````python
print("True  or True  is", True or True)
print("True  or False is", True or False)
print("False or True  is", False or True)
print("False or False is", False or False)
````

One warning about `or` in English. "Bring a passport or a driving licence"
usually means one or the other, not both. Python's `or` is happy with both. It
is false only when everything is false.

````python
day = "saturday"

if day == "saturday" or day == "sunday":
    print("weekend")
else:
    print("weekday")
````

Note the shape of that condition. `day == "saturday" or day == "sunday"` repeats
the `day ==` part on purpose. Writing `day == "saturday" or "sunday"` is a
common attempt and it is wrong: `"sunday"` on its own is a truthy value, so that
condition is true for every day of the week. Exercise 5 shows a tidier way to
write this using `match`.

````python
day = "wednesday"

if day == "saturday" or "sunday":
    print("this says weekend, on a wednesday")
````

## 5. `not` flips one condition

Exercise 2 used `not` on a value. It works the same on a comparison.

````python
password = "sunday42"

has_capital = password != password.lower()

print("has_capital:", has_capital)
print("not has_capital:", not has_capital)

if not has_capital:
    print("add a capital letter")
````

`not` reads best in front of a named condition. `if not has_capital:` is a
sentence. `if not len(password) >= 8:` is a puzzle, and `if len(password) < 8:`
says the same thing better.

## 6. Which operator is applied first

When `and`, `or`, and `not` appear in one condition, Python applies them in a
fixed order: `not` first, then `and`, then `or`.

That means `a or b and c` is read as `a or (b and c)`, which is not what most
people assume on first reading.

````python
print("True or False and False   is", True or False and False)
print("(True or False) and False is", (True or False) and False)
````

Same words, different answers, because the second version says where the
grouping goes.

**Put the brackets in.** They cost nothing, they cannot be misread, and the
reader who follows you does not have to remember a precedence table. A condition
mixing `and` with `or` and no brackets is one of the reliable places to find a
bug in real code.

````python
is_member = False
is_staff = True
spend = 10

# without brackets: reads as is_member or (is_staff and spend >= 50)
print(is_member or is_staff and spend >= 50)

# with brackets: says what it means
print((is_member or is_staff) and spend >= 50)
````

## 7. Python stops as soon as the answer is settled

`and` needs both sides. If the left side is false, the answer is already false,
so Python does not look at the right side at all. `or` behaves the mirror image:
if the left side is true, it stops there.

This is called **short-circuit evaluation**, and it is not a performance
detail. It is what lets you protect a dangerous test with a safe one.

Here is the danger, run on purpose.

````python
orders = 0
total_spent = 90

print(total_spent / orders > 20 and orders > 0)
````

```
ZeroDivisionError: division by zero
```

**`ZeroDivisionError`** is the error type. The cause is that `total_spent /
orders` was evaluated with `orders` holding `0`, because it was written on the
left of the `and`, so Python reached it first.

The guard was there. It was on the wrong side.

````python
orders = 0
total_spent = 90

print(orders > 0 and total_spent / orders > 20)
````

`False`, and no error. `orders > 0` is false, the answer to the `and` is settled,
and the division is never evaluated.

**The test that makes the other one safe goes first.** The same shape protects
you from the `TypeError` in exercise 2:

```python
if basket is not None and len(basket) > 0:
```

Reverse those two and it crashes on the value it was written to survive.

````python
basket = None

print(basket is not None and len(basket) > 0)      # safe
````

## 8. Name the parts, then combine them

A password check written as one expression is correct and unreadable:

```python
if len(pw) >= 8 and pw != pw.lower() and not pw.isalpha() and " " not in pw:
```

Written as named parts, the same rule tells you which rule failed and why.

````python
password = "Sunday 42"

long_enough = len(password) >= 8
has_capital = password != password.lower()
has_non_letter = not password.isalpha()
has_no_space = " " not in password

is_strong = long_enough and has_capital and has_non_letter and has_no_space

print("long_enough   ", long_enough)
print("has_capital   ", has_capital)
print("has_non_letter", has_non_letter)
print("has_no_space  ", has_no_space)
print("is_strong     ", is_strong)
````

Four lines of setup buy you a printable diagnosis. When a user asks why their
password was refused, the one-expression version can only say no.

An `elif` chain turns those parts into a message. It reports the first failing
rule rather than all of them, which is a choice, and exercise 5 comes back to
whether it is the right one.

````python
if not long_enough:
    print("use at least eight characters")
elif not has_capital:
    print("add a capital letter")
elif not has_non_letter:
    print("add a digit or a symbol")
elif not has_no_space:
    print("remove the spaces")
else:
    print("password accepted")
````

## What you learned

- A condition is a value, so it can be given a name, and a named condition reads
  better than the comparison it replaces.
- Five text questions that produce booleans: `len`, `in`, `not in`, `.lower()`
  compared against the original, and `.isalpha()`.
- `and` is true only when both sides are, and it chains.
- `or` is true when either side is, including when both are. `day == "a" or "b"`
  does not mean what it looks like.
- `not` flips one condition, and reads best in front of a named one.
- `not` is applied before `and`, which is applied before `or`. Put the brackets
  in rather than relying on that.
- Short-circuit evaluation: `and` stops at the first false, `or` stops at the
  first true. The test that makes the other one safe goes first.
- Naming the parts turns a verdict into a diagnosis.
