# Errors and debugging

Errors are normal. They are not proof that you are bad at programming; they are part of how programming works. Python tells you what went wrong by raising an exception and printing a traceback. Learning to read that traceback calmly is one of the fastest ways to improve.

A traceback is a record of how the program reached the point of failure. The last line usually names the exception type and message. The lines above it show the call path that led there. Beginners often panic and read from the top, but the bottom is usually where the most important clue lives.

Debugging is the skill of turning confusion into a sequence of small checks. Ask what you expected, what actually happened, and what value or assumption needs to be inspected next.

```python
from __future__ import annotations


def explain_error(name: str) -> str:
    mapping = {
        'TypeError': 'an operation is used on the wrong type',
        'ValueError': 'a value is the right type but wrong content',
        'NameError': 'a name was never defined',
    }
    return mapping.get(name, 'unknown')


if __name__ == "__main__":
    print(explain_error('TypeError'))
```
