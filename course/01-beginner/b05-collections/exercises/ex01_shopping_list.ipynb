## 1. A list holds items in order

Here is the syntax, taken apart.

```
shopping = ["rice", "beans", "oil"]
    │       │   │      │       │  │
    │       │   └──────┴───────┘  │
    │       │        the items    │
    │       └── the opening bracket, and the closing one
    └── an ordinary name, pointing at the whole list
```

Items are separated by commas. The order you write them in is the order the list
keeps, and it keeps it until you change it yourself. `len(x)` gives the number of
items, and `[]` is a list with nothing in it.

````python
shopping = ["rice", "beans", "oil"]

print(shopping)
print("how many:", len(shopping))

empty = []
print(empty, "holds", len(empty), "items")
````

Printing a list shows the brackets and the quotes. That is the list showing you
what it is, which is useful while you work and wrong on a receipt. Section 10
builds the version a person would read.

A list can hold a mixture of types.

````python
mixed = ["rice", 2, 4.5, True]
print(mixed)
````

Whether a list *should* hold a mixture is a separate question. Section 7 shows
what it costs.

## 2. Positions start at zero, and negative positions count from the end

Square brackets after a list ask for the item at a position.

```
shopping = ["rice", "beans", "oil", "salt"]

from the front:      0        1       2       3
from the back:      -4       -3      -2      -1
```

Both rows describe the same four items. Counting from the front starts at zero,
which is the same convention as `range(4)`. Counting from the back starts at
`-1`, because `-0` and `0` would be the same number and the first item already
owns it.

````python
shopping = ["rice", "beans", "oil", "salt"]

print("shopping[0]  ->", shopping[0])
print("shopping[2]  ->", shopping[2])
print("shopping[-1] ->", shopping[-1])
print("shopping[-2] ->", shopping[-2])
````

`shopping[-1]` is the last item with no arithmetic. The alternative,
`shopping[len(shopping) - 1]`, gives the same item and gives you a subtraction to
get wrong.

Positions run out in both directions. On this four-item list, `shopping[4]` and
`shopping[-5]` both raise `IndexError`, which module B04 exercise 5 covered. The
valid front positions are `0` to `len(shopping) - 1`, and the valid back
positions are `-1` to `-len(shopping)`.

A position can also be assigned to, which replaces the item there.

````python
shopping = ["rice", "beans", "oil", "salt"]

shopping[1] = "lentils"
print(shopping)
````

## 3. Slicing takes a run of positions

Two positions separated by a colon give you several items at once.

```
shopping[1:3]
    │     │ │
    │     │ └── stop, not included
    │     └── start, included
    └── the list
```

The stop is not included, which is the `range` rule again, and the same
subtraction tells you the length: `3 - 1` is two items.

Leave a number out and it means "from the beginning" or "to the end".

````python
shopping = ["rice", "beans", "oil", "salt"]

print("[1:3] ->", shopping[1:3])
print("[:2]  ->", shopping[:2])
print("[2:]  ->", shopping[2:])
print("[-2:] ->", shopping[-2:])
print("[:]   ->", shopping[:])

print("the original is untouched:", shopping)
````

A slice always produces a **new list**. Reading a slice never changes the one you
sliced, which is why the last line still shows all four items. `shopping[:]` is a
whole new list holding the same items, and exercise 6 is about why that turns out
to matter more than it looks.

One difference worth knowing now, because it surprises people in both directions:

````python
shopping = ["rice", "beans", "oil", "salt"]

print("a slice past the end stops at the end:", shopping[2:99])
print("and an empty slice is allowed:        ", shopping[3:1])
````

A slice that runs off the end does not complain. A single position that runs off
the end raises `IndexError`. Same list, same overshoot, two different behaviours,
and no rule connects them except that this is how Python is.

The practical consequence: a slice cannot tell you that your numbers were wrong.
If a slice comes back shorter than you expected, or empty, the numbers are the
first thing to check.

## 4. Adding: `append` and `insert`

`.append(item)` puts one item on the end.

`.insert(position, item)` puts one item **at** that position and shifts
everything from there onwards along by one.

````python
shopping = ["rice", "beans", "oil"]

shopping.append("salt")
print("after append:    ", shopping)

shopping.insert(0, "bread")
print("after insert at 0:", shopping)

shopping.insert(2, "eggs")
print("after insert at 2:", shopping)
````

Both change the list **in place**. There is one list here from start to finish,
and it grew. Neither of them hands you a new list, and section 9 is about what
happens when you assume otherwise.

`.append(x)` adds one item. Handing it a list adds that whole list as a single
item, which is rarely what people mean.

````python
shopping = ["rice", "beans"]
shopping.append(["oil", "salt"])

print(shopping)
print("how many items:", len(shopping))
````

Three items became two lists inside one list, and `len` says 3 rather than 4.
`.extend(["oil", "salt"])` is the one that adds them separately, and so is
`shopping + ["oil", "salt"]`, which builds a new list rather than changing this
one.

````python
shopping = ["rice", "beans"]
shopping.extend(["oil", "salt"])
print("extend:", shopping)

together = ["rice", "beans"] + ["oil", "salt"]
print("plus:  ", together)
````

## 5. Removing: `remove` and `pop`

`.remove(value)` deletes the first item **equal to** that value. It works by
value, not by position, and if the value appears twice only the first goes.

`.pop(position)` deletes the item at a position and **hands it back to you**.
`.pop()` with nothing in the brackets takes the last item.

````python
shopping = ["rice", "beans", "oil", "beans", "salt"]

shopping.remove("beans")
print("after remove:", shopping, "  one beans left")

taken = shopping.pop(0)
print("popped:", taken, "  list is now:", shopping)

last = shopping.pop()
print("popped:", last, "  list is now:", shopping)
````

Note the difference in what came back. `.pop` gave you the item it removed, so
`taken` holds `"rice"`. `.remove` gave you nothing at all, which is section 9.

Reach for `.remove` when you know what you want gone and not where it is. Reach
for `.pop` when you know where it is, or when you want the item as well as the
removal.

## 6. Asking a list questions

`in` asks whether a value is present, and produces `True` or `False`. `not in`
is its opposite and reads the way you would say it.

`.count(value)` says how many times a value appears. `.index(value)` says the
position of the first one.

````python
shopping = ["rice", "beans", "oil", "beans", "salt"]

print('"oil" in shopping    ->', "oil" in shopping)
print('"bread" in shopping  ->', "bread" in shopping)
print('"bread" not in list  ->', "bread" not in shopping)
print("count of beans       ->", shopping.count("beans"))
print("first beans is at    ->", shopping.index("beans"))
````

`in` on a list looks at the items one at a time until it finds a match. On five
items that is nothing. On a list of a hundred thousand it is a hundred thousand
comparisons every time you ask, and exercise 3 shows the container that does not
work that way.

`.index(value)` on a value that is not there raises an error rather than
returning `-1`. Section 8 runs the same error from `.remove`.

## 7. Putting a list in order

`.sort()` rearranges the list in place: text alphabetically, numbers by size.
`.sort(reverse=True)` does it backwards.

`.reverse()` flips the current order. It does not sort. On an unsorted list it
gives you the same mess backwards.

`sorted(x)` is different from both: it leaves `x` alone and returns a **new**
list in order.

````python
shopping = ["rice", "beans", "oil", "salt"]

print("sorted(shopping) ->", sorted(shopping))
print("shopping is still ->", shopping)

shopping.sort()
print("after .sort()     ->", shopping)

shopping.reverse()
print("after .reverse()  ->", shopping)
````

Sorting text is not quite alphabetical. Every capital letter sorts before every
lower-case letter, because that is the order the characters are stored in.

````python
names = ["beans", "Apples", "oil", "Salt"]
print(sorted(names))
````

`Apples` and `Salt` come first, and a person looking for `Salt` under S near the
bottom will report it missing. `sorted(names, key=str.lower)` fixes it, and the
intermediate level explains what `key` is doing.

A list that mixes text and numbers cannot be sorted at all. `[3, "one"].sort()`
raises `TypeError`, because Python will not guess whether `"one"` belongs before
or after `3`. That is one of the costs of a mixed list from section 1.

## 8. Removing something that is not there, run on purpose

````python
shopping = ["rice", "beans", "oil"]
shopping.remove("bread")
````

Read the last line.

```
ValueError: list.remove(x): x not in list
```

**`ValueError`** means the type was right and the value was not, which is the
same distinction module B02 drew against `TypeError`. `.remove` wanted a value,
got a value, and that particular value is not in this list.

This is the most common way a list crashes a program, because the item you are
removing usually came from somewhere else: a form, a file, another list. `in`
answers the question before you act on it.

````python
shopping = ["rice", "beans", "oil"]

if "bread" in shopping:
    shopping.remove("bread")
else:
    print("bread was not on the list, nothing to remove")

print(shopping)
````

Deciding what to do when the item is absent is a design decision. Silence,
a message, or a crash are all defensible. Not deciding is what leaves you with
the crash.

## 9. The method that gives you `None`, run on purpose

Now the expensive one. This cell means to keep a tidied copy of the list.

Predict what it prints before you run it.

````python
shopping = ["rice", "beans", "oil", "salt"]

tidy = shopping.sort()

print("tidy is:    ", tidy)
print("shopping is:", shopping)
````

`tidy` is `None`.

**`None`** is a value that means "nothing here". It is what Python hands back
when a piece of code was run for its effect rather than for an answer.

No traceback. No warning. `tidy` exists, it holds a real value, and that value is
useless. The sort did happen, in `shopping`, which is not where you asked for it.

Push one line further and the cost arrives.

````python
print(tidy[0])
````

```
TypeError: 'NoneType' object is not subscriptable
```

**Subscriptable** means "you can put square brackets after it". A list is. A
`None` is not, because there is nothing in it to be at position 0.

Notice where the traceback points: at the line that used `tidy`, not at the line
that set it. In a longer program those two lines can be hundreds apart, in
different files, and the traceback will name the innocent one every time.

**The rule that prevents all of it: a method that changes a list in place returns
`None`.** That is `.append`, `.insert`, `.remove`, `.sort`, `.reverse`, and
`.extend`. They change the list; they do not hand it back.

| You want | Use | It gives you |
|---|---|---|
| the list itself, reordered | `shopping.sort()` | `None` |
| a new list, in order | `sorted(shopping)` | a list |
| the list itself, with an item added | `shopping.append(x)` | `None` |
| a new list with an item added | `shopping + [x]` | a list |

**The symptom is worth memorising: a name holding `None` where you expected a
list means you assigned from an in-place method.**

````python
shopping = ["rice", "beans", "oil", "salt"]

tidy = sorted(shopping)

print("tidy is:    ", tidy)
print("shopping is:", shopping)
````

## 10. A list a person can read

The last job is reporting, and a printed list is not a report. A loop and
`enumerate` from module B04 exercise 5 turn it into one.

````python
shopping = ["rice", "beans", "oil", "salt"]

print("Shopping list")
print("-" * 20)

for position, item in enumerate(sorted(shopping), start=1):
    print(str(position).rjust(2) + ". " + item)

print("-" * 20)
print(len(shopping), "items")
````

`sorted(shopping)` is doing quiet work there. The report comes out alphabetical
and the list itself is untouched, so whoever built the list still has it in the
order they built it. That choice costs one word and is the subject of task 5.

## What you learned

- A list holds items in the order you put them, `len` counts them, and `[]` is
  an empty one.
- Positions start at zero from the front and at `-1` from the back. A position
  past either end raises `IndexError`.
- A slice `[start:stop]` excludes the stop, always produces a new list, and does
  not complain when it runs off the end.
- `.append` adds one item, `.insert(position, item)` shifts the rest along, and
  `.extend` adds several.
- `.remove(value)` deletes by value and gives you nothing back. `.pop(position)`
  deletes by position and hands the item to you.
- `in`, `.count`, and `.index` ask questions, and `in` looks through the items
  one at a time.
- `.sort()` and `.reverse()` change the list in place. `sorted(x)` returns a new
  list and leaves `x` alone. Capitals sort before lower case.
- `.remove` on an absent value raises `ValueError`, and `in` is the guard.
- Every in-place method returns `None`. A name holding `None` where you wanted a
  list is the symptom, and the traceback appears where the `None` was used, not
  where it was made.
