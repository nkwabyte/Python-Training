## 1. What B02 exercise 1 established

Two lines, and the model that explains them.

```
a = 10           a ──────> 10

b = a            a ──┐
                     ├───> 10
                 b ──┘

a = 99           a ──────> 99
                 b ──────> 10
```

`b = a` did not copy anything. It pointed a second label at the same value.
`a = 99` then moved the label `a`, and left `b` where it was.

````python
a = 10
b = a

a = 99

print("a is", a)
print("b is", b)
````

`99` and `10`. The box model predicts the same thing, so nothing here separates
the two models, and this is where most explanations stop.

The next section is where they come apart.

---

## 2. The same two lines with a list, run on purpose

Identical shape. `b = a`, then a change through `b`.

**Write down what you expect `a` to be before you run this.**

````python
a = [10, 20, 30]
b = a

b.append(40)

print("b is", b)
print("a is", a)
````

Both are `[10, 20, 30, 40]`.

There is no traceback to read here. That is the point of this section, and it is
why this failure gets its own notebook. Nothing went wrong as far as Python is
concerned. You asked for exactly this. A cell ran, an answer came out, and the
answer is not the one the code appears to say.

Here is what happened.

```
a = [10, 20, 30]      a ──────> [10, 20, 30]

b = a                 a ──┐
                          ├───> [10, 20, 30]
                      b ──┘

b.append(40)          a ──┐
                          ├───> [10, 20, 30, 40]    one list, changed
                      b ──┘
```

There was never a second list. `b = a` attached a second label to the one that
already existed. `.append` changed that list. `a` points at it, so `a` shows the
change, because there is nothing else for `a` to show.

Section 1 and section 2 are the same two lines. What differs is the third line.
`a = 99` **rebinds**: it moves a label. `b.append(40)` **mutates**: it changes an
object. Two operations that look alike and do opposite things, and almost every
confusing bug involving shared data is somebody who has confused them.

---

## 3. Proving it, rather than believing it

`id(x)` gives a number identifying which object `x` points at. Same number, same
object.

`is` compares those numbers for you. `==` is a different question: it compares
contents.

````python
a = [10, 20, 30]
b = a
c = [10, 20, 30]        # a separate list that happens to look the same

print("a is b?", a is b)
print("a is c?", a is c)
print("a == c?", a == c)

print()
print("id(a) == id(b)?", id(a) == id(b))
````

`a` and `c` are **equal**: they contain the same things. They are not
**identical**: they are two separate lists.

`a` and `b` are both. There is one list and it has two names.

This is the diagnostic. When two names are behaving as one, `a is b` answers it
in a single line, and it answers it definitely.

## 4. Where this costs you: the backup that was not one

Here is the shape it takes in real code. Nobody writes `b = a` and then wonders
what happened. They write this.

````python
shopping = ["rice", "beans", "oil"]

backup = shopping                 # keep the original in case the edits go wrong

shopping.append("salt")
shopping.remove("beans")

print("shopping:", shopping)
print("backup:  ", backup)
````

The backup has the edits in it. There is no copy of the original anywhere. It was
never made, and by the time anybody looks, the information is gone.

The line `backup = shopping` is a comment about intent that the code does not
carry out. It reads as "keep a copy", it means "add a second name", and there is
no warning, no traceback, and nothing in the output of a working program to say
so.

This is the single most common way that beginner Python loses data, and it costs
an afternoon the first time because the cause is nowhere near the symptom.

## 5. Copying a list, three ways

Any of these gives you a genuinely separate list.

| Written | Reads as |
|---|---|
| `list(shopping)` | build a new list from these items |
| `shopping[:]` | a slice of everything, and slices are new lists |
| `shopping.copy()` | copy this |

The slice one is the oldest and you will see it in code you inherit. Exercise 1
said a slice always produces a new list; this is the consequence of that.
`.copy()` is the one that says what it is doing.

````python
shopping = ["rice", "beans", "oil"]

backup_one = list(shopping)
backup_two = shopping[:]
backup_three = shopping.copy()

shopping.append("salt")

print("shopping:    ", shopping)
print("backup_one:  ", backup_one)
print("backup_two:  ", backup_two)
print("backup_three:", backup_three)

print()
print("is it the same object?", shopping is backup_three)
print("does it hold the same things now?", shopping == backup_three)
````

All three backups kept the original three items. The last two lines are the
proof: a different object, and no longer equal, because only one of them changed.

## 6. Rebinding is not mutating

The table that separates the two operations, now for every container you have.

| Written | What it does | Do other names see it? |
|---|---|---|
| `b.append(4)` | changes the list | yes |
| `b[0] = 9` | changes the list | yes |
| `b.sort()` | changes the list | yes |
| `b = [9, 9, 9]` | moves the label `b` | no |
| `b = b + [4]` | builds a new list, moves `b` to it | no |

The last two rows are worth staring at. `b += [4]` on a list behaves like
`.extend` and mutates, while `b = b + [4]` builds a new list. Those two lines
look like the same line and they are not.

````python
a = [1, 2]
b = a
b = b + [3]              # a new list, and b moves to it
print("after b = b + [3]:", "a is", a, " b is", b)

a = [1, 2]
b = a
b += [3]                 # changes the list both names point at
print("after b += [3]:   ", "a is", a, " b is", b)
````

Two characters of difference, opposite consequences, and no error either way.

## 7. Dictionaries and sets do it too

Nothing about this is special to lists. Any container that can be changed
behaves this way, because the reason is about names and objects rather than
about lists.

````python
settings = {"theme": "dark", "size": 12}
saved = settings

settings["size"] = 20

print("settings:", settings)
print("saved:   ", saved)
````

`saved` shows `20`. Same fact, different container.

Dictionaries and sets both have `.copy()`, and `dict(settings)` and
`set(members)` build new ones the same way `list(...)` does.

````python
settings = {"theme": "dark", "size": 12}
saved = settings.copy()

settings["size"] = 20

print("settings:", settings)
print("saved:   ", saved)
print("same object?", settings is saved)
````

## 8. A copy is only one level deep

Now the version that catches people who already know everything above.

Exercise 4 built lists of records: a list whose items are dictionaries. Copy one
of those and see what you get.

````python
students = [
    {"name": "ama", "mark": 72},
    {"name": "kofi", "mark": 55},
]

backup = students.copy()

backup[0]["mark"] = 0                 # change a record through the backup

print("backup:  ", backup)
print("students:", students)
````

Ama's mark is `0` in both.

No traceback again, and this one is worse than section 2 because the copy is
real. `students is backup` is `False`. There genuinely are two lists.

What `.copy()` copied is the **outer** list: a new list, holding the same two
dictionaries. It did not copy the dictionaries, because nothing asked it to.

```
students ──> [ • , • ]
               │   │
               ▼   ▼
             {ama} {kofi}        <- one pair of dictionaries
               ▲   ▲
               │   │
backup ────> [ • , • ]
```

Two lists, two items each, four arrows, and two dictionaries. Changing a
dictionary through either list changes it for both.

This is called a **shallow copy**, and every copy in section 5 is one. Adding a
record to `backup` really does leave `students` alone. Changing a record inside
`backup` does not.

````python
backup.append({"name": "yaa", "mark": 88})

print("backup has  ", len(backup), "records")
print("students has", len(students), "records")
````

## 9. Copying all the way down

Two ways. The first uses only what you have: walk the outer list and copy each
record as you go.

````python
students = [
    {"name": "ama", "mark": 72},
    {"name": "kofi", "mark": 55},
]

backup = []
for student in students:
    backup.append(student.copy())

backup[0]["mark"] = 0

print("backup:  ", backup)
print("students:", students)
````

Ama's mark is `0` in the backup and `72` in the original. Each record was copied,
so there are now four dictionaries rather than two.

The second way is a standard library tool that does this to any depth. `import`
brings in part of the standard library, and a later module covers it properly.

````python
import copy

students = [
    {"name": "ama", "mark": 72},
    {"name": "kofi", "mark": 55},
]

backup = copy.deepcopy(students)

backup[0]["mark"] = 0

print("backup:  ", backup)
print("students:", students)
````

`copy.deepcopy` follows every level and copies everything it finds. That is what
you want when the nesting is deep or you do not know how deep it goes.

It is not free. It walks the whole structure and builds a new one, so on a large
structure copied inside a loop it is a real cost. The loop version above copies
one level and is the right answer when you know one level is all there is.

| You want | Use |
|---|---|
| a separate list of the same items | `.copy()`, `list(x)`, or `x[:]` |
| a separate list of separate records | a loop copying each, or `copy.deepcopy` |
| both names to see each other's changes | `b = a`, on purpose, with a comment |

## 10. What cannot be changed cannot spring this trap

Exercise 3 introduced tuples. Here is the property that makes them worth having.

````python
a = (10, 20, 30)
b = a

b[0] = 99
````

```
TypeError: 'tuple' object does not support item assignment
```

**`TypeError`**, and read what it has done for you. There is no way to write the
bug from section 2 with a tuple. The dangerous operation does not exist, so no
amount of sharing a tuple between names can cause a surprise.

The same is true of numbers, of `True` and `False`, and of text. `b = a` on any
of them is safe, permanently, and that is why section 1 with `a = 10` looked so
harmless.

```
can be changed in place:  list, dict, set        <- aliasing matters
cannot be changed:        int, float, str, bool, tuple   <- aliasing is free
```

That line is the whole rule. When you hand a group of values to another part of a
program and you do not want it changed underneath you, a tuple makes that
impossible rather than merely discouraged.

## 11. When sharing is what you want

Copying by reflex is not the lesson. Sharing is the default in Python because it
is usually right: it costs nothing, and one list that everything points at is one
list to keep correct.

The lesson is that sharing must be a decision. Two names for one list is right
when both names are two views of one thing. It is wrong when one of them is meant
to be a record of how things were.

The question to ask at the moment you write `b = a`: **if this changes, should
the other one change too?** Yes means share and say so in a comment. No means
copy, and say which depth.

## What you learned

- Module B02 exercise 1 said a name is a label. Everything in this notebook is
  that one fact applied to containers.
- `b = a` never copies. It attaches a second label to the same object, and with
  a list that means a change through either name is visible through both.
- The failure has no traceback. It is a wrong answer in a program that runs.
- `id`, `is`, and `==` separate identity from equality, and `a is b` diagnoses a
  shared object in one line.
- The usual disguise is a backup that was never made, which loses the original
  with nothing in the output to say so.
- `list(x)`, `x[:]`, and `x.copy()` all build a separate list.
- Mutating is seen by every name. Rebinding moves one name only, and `b += [4]`
  mutates while `b = b + [4]` rebinds.
- Dictionaries and sets alias the same way, and both have `.copy()`.
- Every one of those copies is shallow. Copying a list of records gives you a
  new list holding the same records.
- Copying each record in a loop goes one level deeper, and `copy.deepcopy` goes
  all the way down at a cost.
- Tuples, numbers, and text cannot be changed in place, so aliasing them is
  free. That is what immutability buys.
- Sharing is often right. It should be a decision, and the question is whether a
  change to one should show up in the other.
