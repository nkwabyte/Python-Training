## Exercise 2: A phonebook you look things up in

A list answers one question well: what is at position 3. That is the wrong
question for most data. Nobody knows which position Ama is at, and if somebody
inserts a name at the front, everyone's position changes and every number you
wrote down is now wrong.

A **dictionary** answers a different question: what is stored under the name
"ama". Positions never enter into it. This is the container most real programs
are mostly made of, and this notebook builds one and then breaks it in the two
ways it breaks.

## 1. A dictionary stores values under keys

Here is the syntax, taken apart.

```
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}
      │      │  │  │       │      │
      │      │  │  │       │      └── a comma between pairs
      │      │  │  │       └── the value
      │      │  │  └── a colon joins one key to one value
      │      │  └── the key
      │      └── curly braces hold the whole thing
      └── an ordinary name
```

Each `key: value` pair is called an **entry** or an **item**. `len` counts the
entries, and `{}` is an empty dictionary.

The usual short name for a dictionary is **dict**, and that is what other
people's code and every error message will call it.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

print(phonebook)
print("how many entries:", len(phonebook))

empty = {}
print(empty, "holds", len(empty), "entries")
````

## 2. Looking up with square brackets

The same square brackets a list uses, with a key inside instead of a number.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

print(phonebook["ama"])
print(phonebook["kofi"])
````

There is no `phonebook[0]`. Position is not how a dictionary finds things, and
`0` would be read as a key you have not got.

Keys are matched exactly. `"Ama"` is a different key from `"ama"`, and `"ama "`
with a trailing space is a third. A dictionary will not do you the favour of
guessing, which is why keys that came from typing are usually cleaned first.

---

## 3. Adding and updating look identical

Assigning to a key stores a value under it. If the key was not there, the entry
is created. If it was, the old value is replaced.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

phonebook["yaa"] = "055 777 8888"        # a new entry
print("after adding yaa: ", phonebook)

phonebook["ama"] = "024 999 0000"        # replaces the old number
print("after updating ama:", phonebook)
````

One line of syntax, two different jobs, and no way to tell from the line which
one it did. That is a real trap and it has a specific shape: a typo in a key you
meant to update does not raise anything at all.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

phonebook["kofi "] = "020 555 6666"      # note the trailing space

print(phonebook)
print("entries:", len(phonebook))
print("kofi is still on:", phonebook["kofi"])
````

Three entries where there should be two, the update went nowhere near Kofi, and
nothing complained. The old number is still the one anybody looking up `"kofi"`
will get.

There is no traceback for this and no way to write one. Section 7 gives you the
check that catches it before you write.

## 4. Asking for a key that is not there, run on purpose

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

print(phonebook["kwame"])
````

Read the last line.

```
KeyError: 'kwame'
```

**`KeyError`** means the key you asked for is not in the dictionary. The message
is the key itself, in quotes, which is more helpful than most errors: it tells
you exactly what was asked for, so you can see the trailing space or the capital
letter you did not mean to type.

This is the correct behaviour, and it is worth saying why. `phonebook["kwame"]`
is a claim that Kwame is in the book. If he is not, then something upstream is
wrong, and stopping at the line that made the false claim is the cheapest place
to find out.

Sometimes the claim is not one you want to make. That is section 5.

## 5. `get` looks up without raising

`.get(key)` returns the value if the key is there, and `None` if it is not. It
never raises `KeyError`.

`.get(key, default)` returns your own value instead of `None`.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

print("get ama:              ", phonebook.get("ama"))
print("get kwame:            ", phonebook.get("kwame"))
print("get kwame with default:", phonebook.get("kwame", "not listed"))

print("the book is unchanged:", phonebook)
````

Two things about that output.

`.get("kwame")` produced `None`, which module B05 exercise 1 introduced as the
value meaning "nothing here". You met it there coming back from `.sort()`.

And the last line matters more than it looks. `.get` does **not** add the key.
Looking something up never creates it, so a lookup cannot quietly grow your
dictionary.

---

## 6. `get` without a default, run on purpose

`get` moved the failure. It did not remove it.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

number = phonebook.get("kwame")
print("ring " + number)
````

```
TypeError: can only concatenate str (not "NoneType") to str
```

**`TypeError`** means the type was wrong. `+` between two pieces of text joins
them, which module B02 covered, and `None` is not a piece of text.

Look at where the traceback points. Not at the lookup, which succeeded on its own
terms, but at the line that used the result. The lookup was two lines away here.
In a real program it can be two files away, and the traceback will still name the
innocent line.

This is the same shape as the `None` from `.sort()` in exercise 1, and the same
lesson: a call that returns `None` on failure hands the failure to whoever uses
the result next.

**Use the default.** `.get(key, default)` is the form worth making habitual,
because it is the form that cannot pass a `None` downstream.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

number = phonebook.get("kwame", "no number on file")
print("ring " + number)
````

So there are three ways to handle a key you are not sure of, and they are three
different decisions rather than three styles.

| Written | When the key is missing | Use it when |
|---|---|---|
| `phonebook[key]` | raises `KeyError` | a missing key means a bug upstream |
| `phonebook.get(key)` | gives `None` | you will test for `None` yourself, next line |
| `phonebook.get(key, d)` | gives `d` | a sensible stand-in exists |

## 7. `in` tests keys

`in` on a dictionary asks about **keys**, not values. This is the check that
catches the trailing-space update from section 3, and the one to write before a
lookup you are not sure of.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444"}

print('"ama" in phonebook          ->', "ama" in phonebook)
print('"kwame" in phonebook        ->', "kwame" in phonebook)
print('"024 111 2222" in phonebook ->', "024 111 2222" in phonebook)
````

The last line is `False`, and that is the one people get wrong. The number is in
there, as a value. `in` did not look at the values at all.

To ask about values, say so.

````python
print('in values ->', "024 111 2222" in phonebook.values())
````

Searching the values goes through them one at a time, the way `in` on a list
does. Searching the keys does not. Exercise 3 explains what makes keys different.

## 8. Removing entries

`del phonebook[key]` removes an entry and gives you nothing back.

`.pop(key)` removes it and hands you the value, the way a list's `.pop` does.
Both raise `KeyError` on a key that is not there, and `.pop(key, default)` is the
form that does not.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444", "yaa": "055 777 8888"}

del phonebook["kofi"]
print("after del:", phonebook)

number = phonebook.pop("yaa")
print("popped:", number, " left with:", phonebook)

gone = phonebook.pop("kwame", "was not there anyway")
print("popping a missing key with a default:", gone)
````

## 9. Walking a dictionary

A `for` loop over a dictionary gives you its **keys**, one at a time.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444", "yaa": "055 777 8888"}

for name in phonebook:
    print(name, "->", phonebook[name])
````

That works, and there is a better way. Three methods say exactly what you want.

- `.keys()` gives the keys, which is what the bare loop above gave you.
- `.values()` gives the values, with no keys attached.
- `.items()` gives both, as a pair per pass.

A pair needs two names to receive it, and two names on the left of the `for` is
the same shape as `enumerate` in module B04 exercise 5. Exercise 3 of this module
explains what a pair actually is.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444", "yaa": "055 777 8888"}

print("keys:  ", list(phonebook.keys()))
print("values:", list(phonebook.values()))

print()
for name, number in phonebook.items():
    print(name.ljust(8) + number)
````

`list(...)` there turns the keys into a list so they print readably. Without it
you get `dict_keys([...])`, which is Python being precise about what it handed
you and is not what you want in output.

`.ljust(8)` pads text on the **right** to a fixed width, so the numbers line up.
It is the mirror of `.rjust` from module B04 exercise 4.

A dictionary keeps entries in the order you added them. That is a promise Python
makes and you can rely on it. It is not sorted order, and if you want that, ask.

````python
phonebook = {"yaa": "055 777 8888", "ama": "024 111 2222", "kofi": "020 333 4444"}

print("insertion order:", list(phonebook))
print("sorted by name: ", sorted(phonebook))
````

`sorted(phonebook)` sorts the **keys**, because that is what a loop over a
dictionary produces. It gives you a list, and it leaves the dictionary alone.

## 10. Keys are unique, and the last one wins

You cannot have two entries under one key. Writing the same key twice does not
raise anything; the later value quietly replaces the earlier one.

````python
phonebook = {"ama": "024 111 2222", "kofi": "020 333 4444", "ama": "055 000 1111"}

print(phonebook)
print("entries:", len(phonebook))
````

Two entries from three pairs, and the first Ama number is gone with no record
that it existed.

Written out by hand like that, it looks like a mistake nobody would make. It is
not a mistake people make by hand. It is what happens when a dictionary is built
in a loop from data that has a duplicate in it, and the symptom is a report whose
total is lower than the file it came from.

Uniqueness is exactly what you want when a key is an account number and it is a
disaster when the key is a surname. Choosing keys is choosing what your program
believes cannot repeat.

## What you learned

- A dictionary stores values under keys, written `{key: value}`, and `len`
  counts the entries.
- Lookup is `phonebook[key]`. Keys match exactly, and there are no positions.
- Assigning to a key adds it or replaces it, with no way to tell which from the
  line, so a mistyped key creates a second entry in silence.
- A key that is not there raises `KeyError`, and the message is the key itself.
- `.get(key)` returns `None` instead of raising, and `.get(key, default)`
  returns your own stand-in. Neither adds the key.
- A `None` from `.get` fails later, somewhere else, usually as a `TypeError`. The
  default is what stops that.
- `in` tests keys. To test values, say `in phonebook.values()`.
- `del` removes an entry, `.pop(key)` removes it and hands back the value, and
  `.pop(key, default)` does not raise.
- A loop over a dictionary gives keys. `.keys()`, `.values()`, and `.items()`
  say which you want, and entries stay in the order you added them.
- Keys are unique. A repeated key keeps the last value and destroys the first
  without saying anything.
