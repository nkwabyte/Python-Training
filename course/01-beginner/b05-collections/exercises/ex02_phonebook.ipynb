# Working with collections

A collection is a value that can hold multiple pieces of data. In beginner Python, the most useful collections are **lists**, **dictionaries**, and **sets**. A list keeps items in order, a dictionary stores key-value pairs, and a set keeps unique items only.

You choose a collection based on the kind of question you need to answer. If order matters, a list is usually a good fit. If you need to look something up by name, a dictionary is often better. If you only care whether an item has appeared before, a set can be the simplest answer.

Learning collections is really about learning how to model information. Once you can choose the right shape for your data, your code becomes easier to read and easier to extend.
