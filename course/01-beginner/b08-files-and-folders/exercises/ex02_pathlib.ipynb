# Files and folders

Programs often need to save work, load saved data, or inspect the filesystem around them. That is where files and folders come in. A file stores data, and a folder helps organize files into a structure you can navigate.

When Python reads or writes a file, it is communicating with the operating system. That means file code can fail for reasons outside your program: the file may not exist, the path may be wrong, or permissions may block access. So file work teaches both useful functionality and careful thinking about errors.

The big beginner habits here are: use clear paths, open files with the right mode, close them safely by using `with`, and think carefully about where your data is being stored.

Create a path object and print its parent and name.


```python
from pathlib import Path


def path_info(folder: str, filename: str) -> str:
    path = Path(folder) / filename
    return f"parent={path.parent}, name={path.name}"


if __name__ == "__main__":
    print(path_info('data', 'notes.txt'))
```
