```python
from __future__ import annotations
import tempfile
from pathlib import Path


def safe_write(path: str, text: str) -> Path:
    tmp = Path(tempfile.mkstemp(prefix='safe_', suffix='.tmp')[1])
    tmp.write_text(text, encoding='utf-8')
    target = Path(path)
    tmp.replace(target)
    return target
```
