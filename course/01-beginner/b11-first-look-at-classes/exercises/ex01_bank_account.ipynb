# A first look at classes

A **class** is a way to group data and behaviour together. Instead of passing a loose collection of values around, you can create an object that stores state and also knows how to act on that state.

Classes become useful when the data and the operations on that data naturally belong together. For example, a bank account has a balance and can deposit or withdraw money. That combination of values and actions is easier to understand when it lives in one object.

At this stage, the goal is not to master advanced OOP. It is simply to learn when a class can make code clearer than a pile of separate functions and dictionaries.

## Task

Create a small bank account object with deposit and balance behaviour.


```python
from __future__ import annotations


class BankAccount:
    def __init__(self, owner: str, balance: float = 0.0) -> None:
        self.owner = owner
        self.balance = balance

    def deposit(self, amount: float) -> float:
        self.balance += amount
        return self.balance

    def withdraw(self, amount: float) -> float:
        self.balance -= amount
        return self.balance


if __name__ == "__main__":
    account = BankAccount('Ada', 100)
    print(account.deposit(50))
```
