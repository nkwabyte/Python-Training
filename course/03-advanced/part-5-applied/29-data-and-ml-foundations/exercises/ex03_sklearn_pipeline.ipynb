### Task 1: Building a Leak-Free Preprocessing Pipeline

Construct `build_classifier_pipeline(numeric_features, categorical_features)` that returns a `Pipeline` containing:
1. `ColumnTransformer` with `StandardScaler` for numeric columns and `OneHotEncoder` for categorical columns.
2. `LogisticRegression` classifier.


```python
# ANSWER 1
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression

def build_classifier_pipeline(num_cols: list[str], cat_cols: list[str]) -> Pipeline:
    preprocessor = ColumnTransformer(
        transformers=[
            ("num", StandardScaler(), num_cols),
            ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols)
        ]
    )
    return Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression())
    ])

```

## Self-Check Harness


```python
import pandas as pd
import numpy as np

def check(passed: bool, msg: str) -> bool:
    status = "PASS" if passed else "FAIL"
    print(f"{status}  {msg}")
    return passed

pipe = build_classifier_pipeline(["age", "fare"], ["city"])

X = pd.DataFrame({
    "age": [25, 45, 30, 50],
    "fare": [10.5, 75.0, 20.0, 100.0],
    "city": ["London", "Paris", "London", "Berlin"]
})
y = np.array([0, 1, 0, 1])

pipe.fit(X, y)
preds = pipe.predict(X)

results = [
    check("preprocessor" in pipe.named_steps, "Task 1: Preprocessor step exists"),
    check("classifier" in pipe.named_steps, "Task 1: Classifier step exists"),
    check(len(preds) == 4, "Task 1: Fit and predict succeeded on sample dataset"),
]
print(f"Summary: {sum(results)}/{len(results)} checks passed.")

```
