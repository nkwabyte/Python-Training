### Task 1: Revenue Aggregation by Category & Region

Write `compute_regional_metrics(df)` that accepts a DataFrame with `['region', 'category', 'revenue', 'quantity']` and returns a grouped summary containing the total revenue, total quantity, and average revenue per unit (`total_revenue / total_quantity`) per `(region, category)` pair.


```python
# ANSWER 1
import pandas as pd

def compute_regional_metrics(df: pd.DataFrame) -> pd.DataFrame:
    summary = df.groupby(["region", "category"], as_index=False).agg(
        total_revenue=("revenue", "sum"),
        total_quantity=("quantity", "sum")
    )
    summary["avg_unit_price"] = summary["total_revenue"] / summary["total_quantity"]
    return summary

```

## Self-Check Harness


```python
def check(passed: bool, msg: str) -> bool:
    status = "PASS" if passed else "FAIL"
    print(f"{status}  {msg}")
    return passed

sample_data = pd.DataFrame({
    "region": ["North", "North", "South", "South"],
    "category": ["A", "A", "B", "B"],
    "revenue": [100.0, 200.0, 150.0, 50.0],
    "quantity": [2, 3, 5, 5]
})

res = compute_regional_metrics(sample_data)

results = [
    check(len(res) == 2, "Task 1: Grouped into 2 region-category pairs"),
    check(res.loc[res['region'] == 'North', 'total_revenue'].values[0] == 300.0, "Task 1: Total revenue computed correctly"),
    check(res.loc[res['region'] == 'North', 'avg_unit_price'].values[0] == 60.0, "Task 1: Average unit price is 300 / 5 = 60.0"),
]
print(f"Summary: {sum(results)}/{len(results)} checks passed.")

```
