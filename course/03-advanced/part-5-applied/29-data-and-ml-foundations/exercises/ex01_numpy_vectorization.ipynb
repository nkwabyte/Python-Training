### Task 1: Vectorized Feature Normalization (Min-Max Scaling)

Implement `min_max_scale(X)` where `X` is a 2D array of shape `(N, D)`. Return the column-wise normalized array where each column scales between `0.0` and `1.0`. Do not use any `for` loops.


```python
# ANSWER 1
def min_max_scale(X: np.ndarray) -> np.ndarray:
    col_min = X.min(axis=0)
    col_max = X.max(axis=0)
    # Guard against zero-division for constant columns
    span = np.where(col_max - col_min == 0, 1.0, col_max - col_min)
    return (X - col_min) / span

```

### Task 2: Pairwise Euclidean Distances (Broadcasting)

Implement `pairwise_distances(A, B)` where `A` has shape `(N, D)` and `B` has shape `(M, D)`. Compute the Euclidean distance matrix of shape `(N, M)` using NumPy broadcasting without `for` loops.


```python
# ANSWER 2
def pairwise_distances(A: np.ndarray, B: np.ndarray) -> np.ndarray:
    # Shape A: (N, 1, D), Shape B: (1, M, D)
    diff = A[:, np.newaxis, :] - B[np.newaxis, :, :]
    return np.sqrt(np.sum(diff ** 2, axis=-1))

```

## Self-Check Harness


```python
def check(passed: bool, msg: str) -> bool:
    status = "PASS" if passed else "FAIL"
    print(f"{status}  {msg}")
    return passed

data = np.array([[10.0, 200.0], [20.0, 100.0], [30.0, 300.0]])
scaled = min_max_scale(data)

pts1 = np.array([[0.0, 0.0], [3.0, 4.0]])
pts2 = np.array([[0.0, 0.0], [1.0, 0.0]])
dists = pairwise_distances(pts1, pts2)

results = [
    check(np.allclose(scaled.min(axis=0), [0.0, 0.0]) and np.allclose(scaled.max(axis=0), [1.0, 1.0]), "Task 1: Normalized columns scale from 0 to 1"),
    check(dists.shape == (2, 2) and np.isclose(dists[1, 0], 5.0) and np.isclose(dists[0, 0], 0.0), "Task 2: Pairwise distance calculation verified"),
]
print(f"Summary: {sum(results)}/{len(results)} checks passed.")

```
